### **What is the risk of writing to a file but never calling close() on it?**

If you write to a file but never call `close()`, two main risks arise:

1. **Data loss:** Python may keep written data in a buffer instead of immediately saving it to the file. Without closing, pending data might not be flushed (saved).
2. **Resource leakage:** The file may remain open, consuming system resources.

**The key point:** `close()` ensures pending changes are saved and releases the file resources.

# **Big Text File Handling**

````python
## Commands Used During Big Text File Handling:

1. f.readline()
2. !cat numbers.txt # Text-File_name
3. flag = 0
4. flag = 1
5. run = numbers.py # CPython-File_name

## Python Code For Big Text File Handling:

f = open("numbers.txt","r")                  # Creates, and opens the file named "numbers.txt" in read mode.
flag = 0                                     # Initializes 0 to flag

while (f.readline() != ''):
    s = f.readline()
    if s != '':
        n = int(s)                          # Conversion : String To Integer
        if (n == 8662807685):
            print("Found!")
            flag = 1
            break                           # Stop The Loop As Soon As It Founds The Number
if (flag == 0):
    print("Not Found!")
````

### **1. Why use f.readline() in a loop instead of reading the entire 230MB file at once?**

**Answer:** Loading the whole file at once can be taxing on system memory, so reading line-by-line avoid this.

Your confusion comes from mixing up **execution order** and **program structure**.
