```python
# Creating Dictionary :

d = {}  # Dictionary Creation

# Filling (Key-Value) Pairs Within Dictionary : ---------------

d["Name"] = "Shivam Kumar"
d["Age"]  = 22
d["Gender"] = "Male"
d["Address"] = "Ballia, U.P"

print(d)
```

```python
text = "Language naturally loops back on itself, weaving familiar words into patterns that anchor our thoughts as we speak, write, and think. Sometimes a single word just fits so perfectly that repeating it—saying it again and again—gives the message a rhythm, a momentum, and a clear point of focus that simply cannot be replaced by any other word. That repetition builds emphasis, and that emphasis creates a steady, memorable cadence, proving that repeating the right word at the right time isn't just repetition for its own sake, but a deliberate choice to make every single sentence resonate."

sl = text.split(" ")   # Splitting The Paragraph Word By Word

l = list(sl)           # Converting The Splitted Words into a List
print(l)               # Printing The List

s = set(l)             # Converting The List into Set
print(s)               # Printing The Set

t = tuple(l)           # Converting The List into tuple
print(s)               # Printing The Tuple

# How To convert into dictionary:

d = {}                # Empty Dictionary Creation

for i in l:           # Picking every single element within the string
  d[i] = 0            # Assigning the picked element as keys within the dictionary and providing value to that key
print(d)              # printing the dictionary


## Note : A dictionary always comes with "Key" and "Value" pairs.

print(d["a"])

```

```python
# Finding the no. of occurence of each element within the dictionary :

d1 = {}                # Empty Dictionary

for i in l:
  d1[i] = d[i] + 1
print(d1)
```

```python
# Finding that element which has the highest number of occurences within the dictionary :

max_element = ""
max = 0

for i in d1:
  if d1[i] > max:
    max = d1[i]
    max_element = i
print(max_element,":", max)
```
