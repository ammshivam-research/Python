### **1. What is a Dictionary?**

After lists (which use square brackets `[]`) and tuples (which use round brackets `()`), Python introduces **dictionaries**, which use **curly brackets `{}**`.

Unlike lists or tuples where items are stored in a simple sequence, dictionary is a data structure where elements are written in **key-value pairs** separated by a colon (`:`):

1. **Left-hand side of the colon:** The **Key** (used to look up information).

2. **Right-hand side of the colon:** The **Value** (the actual data stored).

3. **Uniqueness:** Every key in a dictionary must be **unique** because you use it to access values. However, **values can be duplicated**.

### **2. What Can Be Used as a Dictionary Key?**

Not everything can be a key. The Golden Rule is that a key must be **immutable and hashable** (unchangeable).

* **Yes (Allowed):** Integers, floats, Booleans, and strings.

* **No (Not Allowed):** Lists, sets and other dictionaries because they are **mutable** (can be changed after creation).

     #### **What about Tuples? (Yes and No):**

* **Yes**, because tuples are generally immutable.

* **No**, if the tuple contains a mutable item inside it (like a list).

* **Conclusion:** Only tuples that contain strictly immutable items are allowed to be keys because they are **hashable (unchangable)**. and if they are hashable, then we can access a value easily.

### **3. What Can Be Used as a Dictionary Value?**

* **Anything!** There are **no restrictions** on dictionary values.
* Values can be integers, floats, Booleans, strings, lists, sets, tuples, or even other dictionaries.

### **4. Mutability, Copying, and Functions**

* Dictionaries are **mutable** (changeable), meaning all the properties of mutability that apply to lists also apply to dictionaries.

* **Copying:** If you want to copy a dictionary, you must explicitly use the **`copy()` method**, otherwise you will just create a reference to the original dictionary.

* **Functions:** When you pass a dictionary as an argument to a function, it is passed by reference (changes made inside the function affect the original dictionary).

#####  **UNDERSTANDING PASS-BY REFERENCE IN PYTHON** :

  * **What is an "argument"?**

      An argument is just the variable you put inside the parentheses when you run a function.

      Example: `my_function(my_dict)`

      Here, `my_dict` is the argument. It’s just normal data you are feeding into the function.

  * **What is a "default argument"?**

      A default argument is a backup option you set when creating the function, used only if someone forgets to give input.

      Example: `def my_function(my_dict = {})`
      (That's not what the lecture was talking about).

  * **What does "passed by reference" mean?**

      It just means Python **does not create a duplicate** of your dictionary when you pass it to a function.
      Instead, the function works directly on your **original dictionary**. So if the function adds or deletes something inside that dictionary, your original dictionary gets changed too. That’s all it means!

### **6. The Three Most Important Dictionary Methods**

Python provides built-in methods to work with dictionaries easily:

1. **`d.keys()`**: Returns a collection of all the keys in the dictionary.

2. **`d.values()`**: Returns a collection of all the values in the dictionary.

3. **`d.items()`**: Returns a list-like collection where every element is a **tuple** containing two items: the **key** and its corresponding **value** (an example of how Python uses tuples internally).

```python
### Part 1: Without `.copy()` (They share the same data)

# Original dictionary
d1 = {"name": "Alice", "age": 25}

# Point d2 to d1 without copying
d2 = d1

# Modify d2 by adding a new key
d2["city"] = "New York"

print("d1:", d1)
print("d2:", d2)

# *Notice that `d1` changed too, even though we only edited `d2`!*
```

```python
### Part 2: With `.copy()` (They are completely independent)

# Original dictionary
d1 = {"name": "Alice", "age": 25}

# Create a true separate copy using .copy()
d2 = d1.copy()

# Modify d2
d2["city"] = "New York"

print("d1:", d1)
print("d2:", d2)

# This time, `d1` stays completely safe and unchanged because `d2` is a separate copy.*
```
