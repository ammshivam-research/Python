````python
# Find percentage increased :-----

''' Write a function to calculate the percentage increase from the original value to the new value.

    Args:
        original (float): The original value.
        new (float): The new value.

    Returns:
        float: The percentage increase.

    Examples:
    >>> percentage_increased(50, 75)
    50.0
    >>> percentage_increased(80, 100)
    25.0

'''

def percentage_increased(new, original):
  prcnt_incr = ((new / original) * 100 ) - 100
  return prcnt_incr

new = int(input())
original = int(input())

print(percentage_increased(new, original))
````

````python
# Check if ten digit even number

'''Checks if a number is a 10 digit even number.

    Also account for negative numbers discarding the sign.

    Args:
        n (int): The given number.

    Returns:
        bool : result as True or False.

    Examples:
    >>> is_ten_digit_even(8769473839)
    False
    >>> is_ten_digit_even(928948)
    False
    >>> is_ten_digit_even(9289479278)
    True
    >>> is_ten_digit_even(-9289479278)
    True
    '''

def is_ten_digit_even(n):
  if len(str(abs(n))) == 10 and n % 2 == 0:
    return True
  else:
    return False

n = int(input())
print(is_ten_digit_even(n))
````

````python
# Arithmetic Operations tuple from two integers

'''
    Given a tuple of two integers (a, b), return a tuple containing the
    sum, difference, product, and quotient (integer division) of the two numbers.

    Arguments:
    t: tuple - a tuple of two integers (a, b)

    Return:
    tuple - a tuple containing the sum, difference, product, and quotient

    Example:
    >>> arithmetic_operations((1, 2))
    (3, -1, 2, 0)
    '''

def arithmetic_operations(t: tuple):
  # Using local variables to compute the operations
  sum_val = t[0] + t[1]
  difference = t[0] - t[1]
  product = t[0] * t[1]
  quotient = t[0] // t[1]

  # Returning a tuple with all the results
  return (sum_val, difference, product, quotient)

t = (4, 5)
print(arithmetic_operations(t))
````

````python
# Format Elements in tuple as "second, first"

''' def format_as_second_comma_first(t: tuple) -> str:

    Formats the two elements in a tuple as "second, first".

    Arguments:
    t: tuple - a tuple two elements

    Return:
    str - a formatted string "second, first"

    Example:
    >>> format_as_second_comma_first(('hello', 'python'))
    'python, hello'
    >>> format_as_second_comma_first((1, 2))
    '2, 1'

    '''

def format_as_second_comma_first(t: tuple):
  n1 = str(t[1])
  n2 = str(t[0])
  f_form = ", ".join((n1,n2))
  return f_form

t = ("Hello", True)
print(format_as_second_comma_first(t))
````

## **Notes: Joining Tuple Elements into a String**

### **1. Accessing tuple elements**

```
t = (True, "Shivam")

t[0]  # True
t[1]  # "Shivam"
```

### **2. Converting elements into strings**

```
str(t[0])  # "True"
str(t[1])  # "Shivam"
```

`str()` converts an element into a string.

### **3. Joining strings with a separator**

Syntax:

```
separator.join(iterable)
```

Example:

```
", ".join(("True", "Shivam"))
```

Output:

```
'True, Shivam'
```

- `", "` → Separator (comma followed by a space).
- `("True", "Shivam")` → Iterable containing strings to be joined.
- `.join()` → Combines the strings using the separator.

### **4. Why parentheses?**

```
", ".join(("True", "Shivam"))  # Correct
```

`.join()` accepts **one iterable argument**, not multiple individual arguments.

```
", ".join("True", "Shivam")  # Incorrect
```

A list can also be used:

```
", ".join(["True", "Shivam"])
```

````python
# Even Characters first and Odd Characters reversed

'''def even_first_odd_reversed(s: str) -> str:

    Return a string with the characters in the even indices first
    and the characters in the odd indices reversed next.

    Arguments:
    s: str - the input string

    Return:
    str - modified string

    Example:
    >>> even_first_odd_reversed('abcde')
    'acedb'
    >>> even_first_odd_reversed('python')
    'ptonhy'
    '''
def stringo(n):
    e_str = ""
    o_str = ""
    for i in range(len(n)):
        if i % 2 == 0:
            e_str += n[i]
        else:
            o_str += n[i]

    rv_str = "".join((reversed(o_str)))
    fs = "".join((e_str,rv_str))
    return fs

n = "ShivamKumar"
print(stringo(n))
````
