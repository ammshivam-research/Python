# Python `while` Loops & Variable Reassignment

## **1. Core Concept**

A `while` loop continues to execute as long as its condition evaluates to `True`. To prevent an **infinite loop**, any variable checked in the loop condition **must be updated/reassigned** inside the loop body so that the condition eventually becomes `False`.

---

## **2. Code Comparison**

### 🟢 Correct Implementation

The input value is successfully reassigned to the `brth` variable. When the user inputs `2004`, the condition `brth != 2004` becomes `False`, terminating the loop.

```python
brth = int(input("Enter your birth year: "))

while brth != 2004:
    brth = int(input("Enter once again: "))  # Variable is updated

print("Good")

```

### 🔴 Incorrect Implementation (Infinite Loop)

The new input value is collected via `int(input(...))` but **not stored** in the `brth` variable. Because `brth` never changes, the condition remains permanently `True`, locking the program in an infinite loop.

```python
brth = int(input("Enter your birth year: "))

while brth != 2004:
    int(input("Enter once again: "))  # BUG: Value is discarded, brth never changes

print("Good")

```

---

## **3. Key Takeaway**

* **Rule:** Always ensure your loop control variable is explicitly reassigned or modified inside the loop body.
* **Why it matters:** Failing to store inputs or update counters results in silent freezing or infinite execution blocks in programs.

````python
# Right_code:

brth = int(input("Enter your birth year: "))

while (brth != 2004):
    brth = int(input("Enter once again: "))
print("Good")
````

````python
# Wrong_code:

brth = int(input("Enter your birth year: "))

while (brth != 2004):
    int(input("Enter once again: ")) # The new given value will not be stored in brth. As a result, python will keep looping again and again because the value is not stored in the brth variable which python is using to compare the condition.
print("Good")
````

### **1. `while` Loop Repetition vs. `if` Single Check**

An `if` statement evaluates its condition once, whereas a `while` loop continues executing as long as the condition holds true.


````python
# IF STATEMENT: Executes code at most once
x = 3
if x < 5:
    print("If check passed once:", x)
    x += 1
````

````python
# WHILE LOOP: Repeatedly executes until condition fails
y = 3
while y < 5:
    print("While loop iteration:", y)
    y += 1
````

### **2. Infinite Loops from an Unchanging Condition**

If the variable controlling the loop is never modified inside the block, the condition remains permanently `True`.

````python
count = 5
# WARNING: This loop never terminates because `count` is never decremented
while count > 0:
  print(count)
````

### **3. Off-by-One Considerations After Loop Completion**

When a variable is updated at the end of a loop iteration, its final value after the loop terminates is often one step past the boundary condition.

Imagine you are walking toward a finish line set at step 5.

You take a step, you're at 4 (the loop says "okay, keep going").

You take your next step, you land on 6, which crosses the finish line.

The loop stops because 6 is past the boundary, but your final variable is now sitting at 6, even though your target was 5. Always remember that your counter will often end up one step past where the loop actually finished.

````python
i = 1
while i <= 3:
    i += 1

# Prints 4, not 3, because the loop incremented `i` one last time before exiting
print("Final value of i:", i)
````

### **4. Loop Body Bypassed if Condition is False Initially**

If the condition evaluates to `False` from the very beginning, the code inside the loop block is skipped entirely.

````python
score = 100

while score < 50:
    print("This line will never execute.")
    score += 1

print("Loop was bypassed completely.")
````

### **5. The Risk of `!=` Causing a Skipped Target (Infinite Loop)**

Using strict inequality (`!=`) with increments greater than 1 can cause the variable to leap past the target value, making the condition permanently non-matching.

````python
pos = 0
# WARNING: pos jumps from 9 to 12, entirely skipping 10.
# Because it never equals 10, this runs infinitely.
while pos != 10:
   print(pos)
   pos += 3
````

### **6. Nested Logic Resetting Counter Variables (Hidden Infinite Loops)**

When you place an if statement inside a while loop, you create a nested control structure.

Every time the loop runs its body, it evaluates that inner if condition. If that condition accidentally resets, modifies, or reverses your loop's counter variable, you can break the mathematical progress required to end the loop.

**The Lesson**: Always trace your variables when mixing loops with conditional statements. Make sure an inner if branch cannot undo the forward progress of your loop control variable.

````python
n = 0

while n < 5:
    print(n)
    n += 1       # Normal increment
    if n == 3:
        n = 0    # TRAP: Resets progress back to the beginning!
````

### **7. Guaranteed Termination Conditions (Healthy Loops)**

A "guaranteed termination condition" means your loop is mathematically engineered to stop. For a loop to terminate, two things must happen:

It must have a clear boundary (e.g., countdown > 0).

The code inside must strictly move the variable closer to that boundary on every single run

**The Lesson**: Every loop needs a definitive "countdown" or "count-up" mechanism where the step size (-= 1, += 2, etc.) guarantees the condition will eventually evaluate to False.

````python
countdown = 3

# Guaranteed to terminate because countdown strictly decreases by 1 each time
while countdown > 0:
    print(countdown)
    countdown -= 1

print("Liftoff!")
````

**Question 6** : Which of these while-loop conditions is guaranteed to eventually become false, assuming n starts as a positive integer?



````python
# Option A: The != Leap Trap (Infinite Loop)

n = 3  # Starts as an odd positive integer
while n != 0:
    print(n)
    n = n - 2  # Subtracts 2 each time (3 -> 1 -> -1 -> -3...)
               # Because n skips 0 entirely, the condition (n != 0)
               # remains permanently True, causing an infinite loop
````

````python
# Option B: The Correct Termination (Safely Reaches 0)

n = 3  # Starts as a positive integer
while n > 0:
    print(n)
    n = n - 1  # Subtracts 1 each time (3 -> 2 -> 1 -> 0)
               # Successfully hits 0. When n becomes 0, the condition
               # (0 > 0) evaluates to False, terminating the loop safely.
````

````python
# Option C: Moving Away from the Boundary (Infinite Loop)

n = 5  # Starts as a positive integer
while n < 100:
    print(n)
    n = n - 1  # Subtracts 1 each time (5 -> 4 -> 3 -> 2 -> 1 -> 0 -> -1...)
               # Instead of moving toward 100, it moves further away into
               # negative numbers, keeping n < 100 permanently True.
````

````python
# Option D: Incrementing Away from the Condition (Infinite Loop)

n = 5  # Starts as a positive integer
while n >= 0:
    print(n)
    n = n + 1  # Adds 1 each time (5 -> 6 -> 7 -> 8 -> 9...)
               # Increases n higher and higher, ensuring the condition
               # (n >= 0) is always True and never becomes False.
````
