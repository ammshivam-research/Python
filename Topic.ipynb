## **Finding and Removing the Minimum Element from a Python List Using a User-Defined Function**

````python
l = [-7,-200,3,-90,5]

def min_el(l):
    min = l[0]
    for i in l:
        if i < min:
            min = i
        else:
            continue
    return min

gh = min_el(l)
df = int(gh)
l1 = []
for i in l:
    if i == df:
        l.remove(i)
        l1.append(df)
print("Updated_List :", l)
print("New_List :", l1)
````

##  📘 **Matrix Multiplication & Dot Product in Python**

## **1. What is Matrix Multiplication?**

Matrix multiplication means taking one row from the first matrix and one column from the second matrix, multiplying their numbers, and adding the answers.

## **2. What is Dot Product?**

To perform a matrix multiplication, we use the **dot product.**

Dot product means:

1. Multiply the numbers in the same positions in two lists.
2. Add all the answers together.

**Example:**

\\[ [1,2]\cdot[5,7] \\]

\\[ =(1\times5)+(2\times7)=19 \\]

## **3. Matrix Multiplication Example**

Given:

\\[ A=\begin{bmatrix}1&2\\3&4\end{bmatrix} \quad B=\begin{bmatrix}5&6\\7&8\end{bmatrix} \\]

Result:

\\[ AB=\begin{bmatrix} 19&22\\\ 43&50 \end{bmatrix} \\]

Each element of the result is calculated using the dot product of a row of A and a column of B.


## **4. Understanding the Three Loops**

Let's consider two matrices:

\\[ A = \begin{bmatrix}1 & 2\\3 & 4\end{bmatrix} \\]

\\[ B = \begin{bmatrix}5 & 6\\7 & 8\end{bmatrix} \\]

Our goal is to multiply A and B.

| Variable | Meaning                                           | Example                  |
| -------- | ------------------------------------------------- | ------------------------ |
| `i`      | Selects a row from A                              | `i = 0` selects `[1, 2]` |
| `j`      | Selects a column from B                           | `j = 0` selects `[5, 7]` |
| `k`      | Selects corresponding elements for multiplication | `1 × 5`, then `2 × 7`    |

### Visual Example

Step 1: `i = 0` → Select the first row of A.

\\[ A = \begin{bmatrix}\boxed{1} & \boxed{2}\\3 & 4\end{bmatrix} \\]

Step 2: `j = 0` → Select the first column of B.

\\[ B = \begin{bmatrix}\boxed{5} & 6\\\\\boxed{7} & 8\end{bmatrix} \\]

Step 3: `k` → Multiply corresponding elements and add them.

\\[ (1 \times 5) + (2 \times 7) = 19 \\]

The first result is 19.

Remember:

- `i` = Which row?
- `j` = Which column?
- `k` = Which elements should I multiply and add?

## **5. Why Must the Number of Columns in A Equal the Number of Rows in B?**

We can multiply two matrices only when:

Number of columns in A = Number of rows in B.

### Example 1: Multiplication is possible

Matrix A has 2 rows and 3 columns (2 × 3).

\\[ A = \begin{bmatrix}1&2&3\\4&5&6\end{bmatrix} \\]

Matrix B has 3 rows and 2 columns (3 × 2).

\\[ B = \begin{bmatrix}1&2\\3&4\\5&6\end{bmatrix} \\]

- A has 3 columns.
- B has 3 rows.

Therefore, multiplication is possible.

\\[ (2 \times \boxed{3})(\boxed{3} \times 2) \\]

Result size = 2 × 2

### Example 2: Multiplication is NOT possible

Matrix A = 2 × 3

Matrix B = 2 × 2

- A has 3 columns.
- B has only 2 rows.

Therefore, multiplication is not possible.

Remember: The inner numbers must match. The outer numbers tell you the size of the resulting matrix.

## **6. Matrix Multiplication vs Element-by-Element Multiplication**

These are two different operations.

### A. Element-by-element multiplication

Multiply the numbers in the same positions.

\\[ \begin{bmatrix}1&2\\3&4\end{bmatrix} \odot \begin{bmatrix}5&6\\7&8\end{bmatrix} = \begin{bmatrix}5&12\\21&32\end{bmatrix} \\]

For example:

- 1 × 5 = 5
- 2 × 6 = 12
- 3 × 7 = 21
- 4 × 8 = 32

### B. Matrix multiplication

Here, we multiply rows by columns and add the products.

\\[ AB = \begin{bmatrix}19&22\\43&50\end{bmatrix} \\]

The results are different because the multiplication methods are different.

## 4. What Is the Dot Product?

Dot product means multiplying corresponding elements of two sequences and adding all the results.

### Example

\\[ [1,2,3]\cdot[4,5,6] \\]

Step 1: Multiply corresponding elements.

- 1 × 4 = 4
- 2 × 5 = 10
- 3 × 6 = 18

Step 2: Add the results.

\\[ 4+10+18=\boxed{32} \\]

Therefore, the dot product is 32.

## 5. Relationship Between Dot Product and Matrix Multiplication

Matrix multiplication uses multiple dot products to calculate the elements of the resulting matrix.

Let's understand with an example:

\\[ A=\begin{bmatrix}1&2\\3&4\end{bmatrix} \\]

\\[ B=\begin{bmatrix}5&6\\7&8\end{bmatrix} \\]

| Result position           | Dot product       | Answer |
| ------------------------- | ----------------- | ------ |
| First row, first column   | (1 × 5) + (2 × 7) | 19     |
| First row, second column  | (1 × 6) + (2 × 8) | 22     |
| Second row, first column  | (3 × 5) + (4 × 7) | 43     |
| Second row, second column | (3 × 6) + (4 × 8) | 50     |

Therefore:

\\[ AB=\begin{bmatrix}19&22\\43&50\end{bmatrix} \\]

## 6. Final Takeaways

1. Dot product: Calculates one number by multiplying corresponding elements and adding their products.
2. Matrix multiplication: Uses multiple dot products to create a new matrix.
3. Three loops:
   - `i` selects a row.
   - `j` selects a column.
   - `k` multiplies corresponding elements and adds their products.

## **Python `@` Operator**

The `@` operator is used to perform matrix multiplication in Python.

It works with supported array types, such as NumPy arrays.

Remember:

- `@` → Matrix multiplication.
- `*` → Element-by-element multiplication (for NumPy arrays).

````python
# Example :

import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

C = A @ B

print(C)
````

````python
# MATRIX MULTIPLICATION EXAMPLE :


A = [
    [1, 2],
    [3, 4]
]

B = [
    [5, 6],
    [7, 8]
]

# Create an empty result matrix
C = [
    [0, 0],
    [0, 0]
]

for i in range(2):
    for j in range(2):
        for k in range(2):
            C[i][j] += A[i][k] * B[k][j]

print("Result:")
print(C)
````

````python
# Generator

def square(limit):
    x = 0
    while x < limit:
        yield x * x
        yield x * x * x
        x += 1

a = square(5)

print(next(a),next(a))
print(next(a), next(a))
print(next(a), next(a))
print(next(a), next(a))
print(next(a), next(a))
````

## **map() function**

**Meaning**: The map() function applies a function to every element of a list.

````python
## Example 1: Multiply every number by 2

numbers = [1, 2, 3, 4, 5]

# Multiply each number by 2
result = map(lambda x: x * 2, numbers)

# Convert the map object into a list
result = list(result)

print(result)

# Output: [2, 4, 6, 8, 10]
````

````python
## Example 2: Using a normal function instead of lambda

numbers = [1, 2, 3, 4, 5]

# Define a function
def square(x):
    return x ** 2

# Apply square() to every element
result = map(square, numbers)

# Convert the result into a list
print(list(result))

# Output: [1, 4, 9, 16, 25]
````

## **filter() function**

**Meaning**: The filter() function selects only those elements that satisfy a given condition.

Think of it like a security guard:
- Checks every element.
- If the element satisfies the condition, it passes.
- Otherwise, it gets rejected.

````python
## Example 1: Find even numbers

numbers = [1, 2, 3, 4, 5, 6, 7, 8]

# Check whether a number is even
result = filter(lambda x: x % 2 == 0, numbers)

# Convert the filter object into a list
result = list(result)

print(result)

# Output: [2, 4, 6, 8]
````

````python
## Example 2: Using a normal function instead of lambda

numbers = [10, 15, 20, 25, 30, 35]

# Define a function
def greater_than_20(x):
    return x > 20

# Select numbers greater than 20
result = filter(greater_than_20, numbers)

# Convert the result into a list
print(list(result))

# Output: [25, 30, 35]
````
