## **Key Concepts Explained: Mutability, Homogeneity, and Order**

### **1. Mutability (Mutable vs. Immutable)**

**What it means**: Whether an object’s internal value can be changed after it has been created, without creating a brand new object in computer memory.

**Mutable (Can change)**: You can modify, add, or delete items inside the same object.

**Example**: A list is mutable because you can change my_list[0] = 99 without changing where the list lives in memory.

**Immutable (Cannot change)**: Once created, its value is locked forever. If you try to modify it, Python actually creates a brand-new object behind the scenes.

**Example**: A string is immutable. If you change a string, Python replaces it with a completely new string in memory.

### **2. Homogeneity (Homogeneous vs. Heterogeneous)**

**What it means**: Whether a data structure can hold only one specific data type or a mixture of different data types.

**Homogeneous (Uniform / Same type)**: Every single element inside the collection must be of the exact same data type.

**Example**: A strict string is homogeneous because every element inside it is a character (str).

**Heterogeneous (Mixed / Different types)**: The collection can mix different types together freely.

**Example**: A Python list can hold an integer, a string, a boolean, and another list all at the same time: [42, "hello", True, [1, 2]].

### **3. Order (Ordered vs. Unordered)**

**What it means**: Whether the items in a data structure maintain a strict, predictable sequence where every item has a fixed position (index) that you can access.

**Ordered**: Items stay in the exact position you put them in, and you can access them using numbers like index 0, 1, 2.

**Examples**: Lists, tuples, and strings.

**Unordered**: The items do not have a guaranteed or fixed position. They might shift around in memory, meaning you cannot access them using standard numerical indices.

**Examples**: Sets (which rearrange items based on hash values for fast searching).

````python
# Solution_1 :

print(tuple((num,)))
print(type(tuple((num,))))

# Solution_2 :

print(tuple(str(num)))
print(type(tuple(str(num))))

````

## **2. Float Data Type (float)**

Whenever we talk about a float data type, we are talking about the decimal numbers.

**Characteristics:**

**Precision Limit:** Floats have limited precision (usually about 15–17 decimal digits of accuracy). i.e in the RAM of our computer, we can store decimal values with upto 15 - 17 decimal digits of accuracy.

**Precision Gotcha:** Due to how computers store binary fractions, floating-point arithmetic can occasionally lead to tiny rounding errors (e.g., 0.1 + 0.2 equals 0.30000000000000004 rather than 0.3).



## **Why 0.1 + 0.2 Doesn't Equal 0.3 in Python?**

To understand why `0.1 + 0.2` doesn't quite equal `0.3` in programming, we have to look at how computers think versus how humans think. It comes down to a translation problem between **Base-10 (Decimal)** and **Base-2 (Binary)**.

---

### 1. The Root Cause: How Computers Store Numbers

Humans use the **decimal system (Base-10)** because we have 10 fingers. Computers use the **binary system (Base-2)** because they are built from electronic switches that only understand two states: `0` (off) and `1` (on).

* In Base-10, fractions like $\frac{1}{2}$ are written as `0.5`, and $\frac{1}{4}$ are written as `0.25`.
* Fractions like $\frac{1}{3}$ cannot be written cleanly in decimal; they result in a repeating decimal: `0.333333...` which goes on forever. If you stop writing it down, you lose a tiny bit of precision.

The exact same thing happens in binary, but for different fractions. In binary (Base-2):

* Powers of two ($1/2$, $1/4$, $1/8$, $1/16$) terminate cleanly.
* Numbers like **$0.1$** and **$0.2$** are actually **infinite repeating fractions** in binary, just like $\frac{1}{3}$ is in decimal.

---

### 2. The Floating-Point Approximation

Because computer memory is finite (standard Python floats use 64 bits), Python cannot store an infinite repeating binary fraction. It has to cut it off (approximate it) at a certain point.

* The true binary representation of `0.1` is actually something like `0.00011001100110011001101...` (repeating forever).
* When Python cuts this off to fit into 64 bits, the stored number isn't *exactly* `0.1`. It's a tiny microscopic fraction off—usually something like `0.10000000000000000555...`.
* Similarly, `0.2` is stored as an approximation slightly different from exact `0.2`.

---

### 3. What Happens During `0.1 + 0.2`?

When you add the two approximate numbers together, the tiny inaccuracies add up as well:

$$\text{Stored } 0.1 + \text{Stored } 0.2 = 0.30000000000000004$$

Python isn't "broken"—it is simply showing you the exact result of the two imperfectly stored approximations added together.

---

### How Do You Handle This in Python?

Because this is a hardware-level limitation baked into almost all modern programming languages (not just Python), developers use a few standard workarounds:

1. **Rounding (`round()`):** If you just want to display the number cleanly to a user, you can round it:
```python
print(round(0.1 + 0.2, 1))  # Output: 0.3

```


2. **The `decimal` Module:** If you need absolute financial or scientific precision where rounding errors are unacceptable, Python has a built-in `decimal` module that does math the way humans do (in Base-10):
```python
from decimal import Decimal

print(Decimal('0.1') + Decimal('0.2'))  # Output: 0.3

```

## **3. Type Conversion (Casting)**

### Converting float to an integer **(int())**: Truncates (Cut off) the decimal part. (rounds toward zero).

## **4. Operations Between int and float**

When you try to perform arithmetic operations between an integer and a float, Python automatically converts (or casts) the integer data type into a float data type, and the resulting data type is always a float data type.

````python
result = 3 + 4.5    # 3 is treated as 3.0
print(result)       # Output: 7.5
print(type(result)) # Output: <class 'float'>
````

**Note:** The division operator (/) always returns a float, even if both operands are integers and divide evenly (e.g., 4 / 2 results in 2.0).
