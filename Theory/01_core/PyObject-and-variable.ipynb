# **Variables and Literals in Python**

## **Variable**

A **variable** is simply a label that we assign to the memory address of the specific part of the **memory array banks**, a *memory location* within the memory chips of **RAM**(Random Access Memory) in a computer where data is explicitly stored in **electric form** within the electrical transistors, to fetch or access the data whenever we want. Writing the memory address is not feasible, because they are in either decimal or hexadecimal format.

A **variable** is not a data structure, but a **label** where values stored within it can be changed. It can be used for either mutable or immutable object.

In Python, you **do not need** to declare the type of variable explicitly unlike SQL; it is created the moment you assign a value to it using the **`=`** operator.

## **Literal**

A literal is not considered to be a data structure, but the data itself. and data can be found in many formats.

**Data Formats**:
1.   string = "Hello"

4.   bool   = True

so i.e "Hello" is a **literal**. 2 is a **literal**. 4.33 is a **literal**. True is a **literal**.


## **OBJECT IN PYTHON**

In Python, **everything is an object**.

Whether it is a simple number (`5`), a piece of text (`"hello"`), a function, a module, or a complex data structure (like a list), it is treated as an object.

At its core, an object is a distinct "box" of data in your computer's memory. This object or the object box contains the type, and the value of the data only, but not the memory address of the object.

but whenever we create an object, it always has **three defining characteristics or things**:

---
### **The Three Characteristics of an Object**

1. **Memory Address**

    This is the object's address in memory. Once an object is     created, its memory address never changes.

    You can check an object's memory address using the built-in `id()` function.

    For example, if you have stored the integer data within a object and label the object using variable named "var", then to find out the memory address of that object, you need to use this syntax in IDE: `id(var)` or `hex(id(var))`


2. **Type (`type`)**

    This defines what kind of object it is (e.g., `int`, `str`, `list`, `dict`).
    
    The type determines what operations the object can perform and whether it is **mutable or immutable**.
    
    You can check an object's type using `type()`.


3. **Value**

    This is the actual data stored inside the object (e.g., the number `42`, or the items inside a list `[1, 2, 3]`).
    
    This is what can either change (if it's mutable) or stay locked (if it's immutable).

````python
### A Quick Example in Python

# Let's look at how this works in code:

x = 42

print(type(x))          # Output: <class 'int'> (Its Type)
print(id(x))            # Output: 10464072 (Its Memory Address / Identity)
print(hex(id(x)))       # Output: 0x9fab48 (Its Memory Address / Identity)
print(x)                # Output: 42 (Its Value)
````

## **Mutability & Immutability**

Mutability is a property of the object.

Mutability describes whether the data stored inside the PyObject can be changed?

**Immutable object**: The data stored inside the PyObject cannot be changed. If you want a different value, you have to throw the old object away and make a new one by replacing the previous one. or just create a new object.

**Mutable object**: The data stored inside the PyObject can be changed. The object stays in the exact same memory location, but its internal contents or data are updated.

### **QUICK COMPARISON TABLE**

| Category | Data Types | Can be modified in place? |
| :--- | :--- | :--- |
| **Immutable** | `int`, `float`, `str`, `tuple`, `bool` | ❌ No |
| **Mutable** | `list`, `dict`, `set` | ✔️ Yes |




````python
# ==========================================
# 1. IMMUTABLE (Cannot change in place)
# ==========================================

# Create a string object
text = "hello"
print("Original ID:", id(text))  # e.g., 140123

# "Modifying" the string
text = text + " world"

# Look at the ID now:
print("New ID:     ", id(text))  # DIFFERENT ID!
# Python didn't change "hello". It threw it away and built a brand-new object.
````

````python
# ==========================================
# 2. MUTABLE (Can change in place)
# ==========================================

# Create a list object
my_list = [1, 2, 3]
print("Original ID:", id(my_list))  # e.g., 987654

# Modifying the list
my_list.append(4)

# Look at the ID now:
print("New ID:     ", id(my_list))  # EXACT SAME ID!
# Python didn't make a new list. It changed the exact same list in memory.
````


### **Variable Naming Rules in Python**

| Rule | Allowed / Valid | Not Allowed / Invalid | Reason |
| :--- | :--- | :--- | :--- |
| **Starting Character** | `score`, `_temperature` | `2nd_place`, `@score`, `#price` | Must start with a letter or an underscore (`_`). **Never** a number or a symbol. |
| **Allowed Characters & Symbols** | `user_score_1`, `total_amount` | `user-score`, `user.name`, `price$`, `total amount` | Only letters, numbers (`0-9`), and underscores (`_`) are allowed. **No symbols or spaces** anywhere inside. |
| **Case Sensitivity** | `age`, `Age`, `AGE` | N/A | Python is case-sensitive; these are treated as 3 completely different variables. |
| **Reserved Words** | `class_name`, `my_def` | `class`, `def`, `if`, `for` | Reserved Python keywords cannot be used as variable names. |

````python
# ==========================================
# VALID VARIABLE NAMES
# ==========================================

score = 100                 # Starts with a letter
_temperature = 25           # Starts with an underscore
user_score_1 = 42           # Uses letters, numbers, and underscores
Age = 30                    # Uppercase letter (valid, treated differently than 'age')
age = 25                    # Lowercase letter (valid, treated differently than 'Age')
class_name = "Math"         # Appending an underscore avoids colliding with reserved keywords
````

````python
# ==========================================
# INVALID VARIABLE NAMES (These cause SyntaxErrors)
# ==========================================

# 2nd_place = 10            # INVALID: Cannot start with a number
# @score = 100              # INVALID: Cannot start with a symbol
# user-score = 50           # INVALID: Cannot contain hyphens or symbols inside
# user.name = "Alice"       # INVALID: Cannot contain dots inside
# total amount = 10         # INVALID: Cannot contain spaces inside
# class = "Math"            # INVALID: Cannot use reserved Python keywords
````

## **How to delete a variable?**

**Meaning** : When we say that we want to delete a variable, i.e we want to break the connection between the label that we assigned to the PyObject and the actual PyObject.

### **Breaking the connection will lead to what?**

If we break the connection between the label that we assigned to the PyObject and the actual PyObject, and after doing that if we didn't try to assign any label to the actual PyObject, then the python garbage collector will remove the actual PyObject from the memory.

### **But how can we delete a variable?**

To delete a variable in Python, you use the **`del`** keyword.

When you use `del` Keyword, Python removes the variable name and breaks the connection to the object in memory.

### **When would you use `del`?**

* To free up memory when working with very large data structures (like massive lists or dataframes) that you no longer need.
* To clean up temporary variables in a script or interactive session.

````python
### Code Example

# 1. Create a variable
x = 42
print(x)  # Output: 42

# 2. Delete the variable
del x

# 3. Trying to use it now will cause an error
print(x)  # NameError: name 'x' is not defined
````

# **Dynamic Typing in Python**

In Python, a variable isn't assigned to just one data type; instead, it can be assigned to any data type or data structure that is available in python.

Variable names act as references or pointers, replacing the need for memory address to access the data from the memory in a more convinient way.

### **Successive(Dynamic) Reassignment (Changing types over time)**

In Python, you can take the exact same variable name and point it to a completely different data type or data structure later in your code.

````python
### Dynamic(Successive) Reassignment : ---------------------------------------

# 1. First, x points to an integer object
x = 10
print(type(x))  # <class 'int'>

# 2. Reassign x to point to a string object
x = "hello"
print(type(x))  # <class 'str'>

# 3. Reassign x again to point to a list object
x = [1, 2, 3]
print(type(x))  # <class 'list'>
````

## **What does variable independence actually mean?**

**Explanation**

In Python, variables are like independent sticky-note labels placed on any PyObject within the memory.

### **How can you explain variable independence when we have a mutable object?**

**Meaning**: If we have `list`, `dictionary`, and `sets` data structure as mutable PyObject, how can you explain variable independence through it?

*To understand the above question, first we should look at the code examples, and then we try to figure out what is really happening?*


````python
## CODE EXAMPLE : ---------------------------------------------------------

l = [1, 2, 3]
# Created our first variable named "l", and assigned the variable "l" to our newly created List PyObject : [1,2,3]

l1 = l
# Created a new variable named "l1", and assigned the variable "l1" to the same List PyObject [1,2,3] as our 1st variable "l" has been assigned.

l.append(100)

print("l is:", l)      # Output : [1, 2, 3, 100]
print("l1 is:", l1)    # Output : [1, 2, 3, 100]


# NOTE : WHAT HAPPENED?

## 1. So we first created our first variable named "l", and assigned the variable "l" to our newly created List PyObject : [1,2,3]
## 2. After that, we created a new variable named "l1", and assigned the variable "l1" to the same List PyObject [1,2,3] as our 1st variable "l" has been assigned.
## 3. Now, our List PyObject [1,2,3] has two assigned variables : l, l1
## 4. Meaning : Instead of using memory address to access our newly created List PyObject [1,2,3], we are going to use l or l1 variable to access it.
## 5. Now we have tried to modify our List PyObject by adding a new element or value to the List PyObject.
## 6. Once we add the element 100 to our List PyObject, what we will observe is that our List Pyobject has modified into : [1,2,3,100]
## 7. i.e by using variable "l" or "l1", we can access the modified version of our List PyObject [1,2,3] : [1,2,3,100]

````

So based on above observation, 1 thing is clear and that is : When we are dealing with mutable data structure PyObject like Lists, Sets, and dictionaries, then the concept of variable independence are not allowed automatically.

But if it is not allowed automatically then how is it allowed? How can a list, set, or dictionary as a mutable data structure PyObject allows the variable independence in Python?

````python
# True Independence (Using .copy())

## Meaning : If you want them to be completely independent, you must copy the data into a brand-new box:

a = [1, 2, 3]
b = a.copy()   # Creates a separate, independent copy of the list

a.append(4)

print("a:", a)  # Output: [1, 2, 3, 4]
print(id(a))

print("b:", b)  # Output: [1, 2, 3]     <-- 'b' stays safe and independent!
print(id(b))

### NOTE : Two distinct variables can be assigned to the same type of data within the PyObject, with two distinct memory addresses is an example of "variable independence."
````

### **How can you explain variable independence when we have an immutable PyObject?**

**Meaning**: If we have `int`, `float`, `str`, `bool`, and `tuple` data type or structure, how can you explain variable independence through it?

As we have seen in the above case with mutable object that variable independence does not happen automatically, but manually. but when we deal with the immutable object, modification of the immutable object in any way will create variable independece automatically.

So with immutable objects, variable independence happens automatically. We do not need a `.copy()` function to make the variable independence happen like when we deal with mutable objects.

plus there isn't any type of function exist in python for immutable objects to create variable independence. because it is unnecessary as we already seen the variable independence happens automatically when we try to modify the PyObject.


### **When we say modifying the PyObject, what is the meaning of it?**

When people casually say "modifying an immutable variable" (like changing a string or a number), what they actually mean is reassigning the variable to a brand-new value within a brand new PyObject.

#### **Here's the code example :**

```
x = 5
x = x + 1  # It looks like we are modifying x
```
*Behind the scenes, Python does not take the number 5 in memory and change it to 6. Instead, here are the steps Python actually takes:*

1. It looks at the current value of x (which is 5).

2. It calculates 5 + 1, which equals 6.

3. It creates a brand-new object in memory for 6.

4. It moves the label x so it now points to 6.

5. Value 5 gets erased by python garbage collector as we know there is no variable pointing to it now.



````python
## CODE EXAMPLE : -----------------------------------------------------

# 1. 'a' points to the string "hello"
a = "hello"

# 2. 'b' points to the exact same string
b = a

print("Before reassignment:")
print("a:", a)  # Output: hello
print(id(a))    # Memory Address 1
print("b:", b)  # Output: hello
print(id(b))    # Memory Address 2

### NOTE :
# 1. At this point, we have 1 string PyObject stored with the value "hello", and two variables assigning to that string PyObject : a, b
# 2. And our newly created string PyObject has 1 memory address as well.

# 3. Reassign 'a' to a new string
a = "world"

print("\nAfter reassigning 'a':")

print("a:", a)  # Output: world (moved to a new object)
print(id(a))    # Memory Address 1
print("b:", b)  # Output: hello (stayed completely safe and independent!)
print(id(b))    # Memory Address 2

### NOTE :
# 1. Now we have 2 string PyObject stored with the values: "world" , "hello"
# 2. and two variables are assigned seperately with each distinct string PyObject.
# 3. And we have two distinct memory addresses as well for each string PyObject.
````

### **SUMMARY**

**With mutable objects** (like lists): You can change the contents inside the existing box without moving the label (e.g., a.append(4)).

**With immutable objects** (like numbers/strings): You cannot change the box. If you want a new value, you have to throw away the old box and point to a completely new one.

## **Key Concepts of Multiple Assignment and Variable Swapping**

| **Concept** | **How It Works** | **Example / Important Point** |
|---|---|---|
| **Multiple Assignment** | Assigns values to multiple variables in one line. | `a, b = 1, 2` assigns `1` to `a` and `2` to `b`. |
| **Variable Swapping** | Exchanges the values of two variables without needing a third variable. | `x, y = y, x` swaps the values of `x` and `y`. |
| **Unpacking Count** | Python checks whether the number of values matches the number of variables. | `a, b = 1, 2, 3` raises a `ValueError` because there are two variables but three values. |

### **Multiple Assignment in Python**

Multiple assignment allows you to assign values to more than one variable in a single line of code. Instead of writing separate lines for every single variable, Python lets you group them efficiently.

**There are two main ways to use multiple assignment in Python:**

1. **Assigning Different Values to Multiple Variables** (Unpacking)

    You can assign multiple values to multiple variables at once by separating them with commas. Python matches them up positionally (left to right).

````python
# Assigns 10 to x, 20 to y, and "Hello" to z
x, y, z = 10, 20, "Hello"

print(x)  # Output: 10
print(y)  # Output: 20
print(z)  # Output: Hello
````

**Note**: The number of variables on the left must match the number of values on the right, otherwise Python will raise a **ValueError**.

````python
# Assigns 10 to x, 20 to y, and "Hello" to z
x, y = 10, 20, "Hello"

print(x)  # Output: 10
print(y)  # Output: 20
````

2. **Assigning the Same Value to Multiple Variables** (Chained Assignment)

    If you want multiple variables to point to the exact same value or object, you can chain them together using multiple `=` signs.

````python
# Assigns the integer 100 to a, b, and c simultaneously
a = b = c = 100

print(a)  # Output: 100
print(b)  # Output: 100
print(c)  # Output: 100
````

### **Bonus: The Classic Python Swap Trick**

One of the coolest practical uses of multiple assignment is swapping the values of two variables without needing a temporary third variable.

````python
a = 5
b = 10

# Swap their values instantly using multiple assignment
a, b = b, a

print("a is now:", a)  # Output: 10
print("b is now:", b)  # Output: 5
````
