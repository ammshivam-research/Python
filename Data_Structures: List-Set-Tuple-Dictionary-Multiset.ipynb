## **2D List (Matrix)**

Think of 2D data like a grid or a spreadsheet with rows and columns.

A normal list like [1, 2, 3] is just a single straight line (1D). But when you put multiple lists inside one big list, you get a grid of rows and columns (2D).

To grab any number, you just need two coordinates: the row number and the column number.

#### **Visual Table (with Row and Column Indexes)**

|  | Column 0 | Column 1 | Column 2 |
| --- | --- | --- | --- |
| **Row 0** | `1` | `2` | `3` |
| **Row 1** | `4` | `5` | `6` |
| **Row 2** | `7` | `8` | `9` |


#### **How to Access Elements**

To grab any item, use two brackets: `grid[row_index][column_index]`.

```python
grid = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Get the number at Row 0, Column 2
print(grid[0][2])  # Output: 3

# Get the number at Row 1, Column 1 (the middle one)
print(grid[1][1])  # Output: 5

# Get the number at Row 2, Column 0
print(grid[2][0])  # Output: 7

```

### **Accessing or checking if an item exists within the set using (in) operator**

Accessing (checking if an item exists using `in`) is **much faster in a set** than in a list.

### **Why Sets are Faster**

* **List ($O(n)$):** A list checks items **one by one** from start to finish. If the item is at the end or not there at all, it has to look through every single element.

    Like looking for a name in a long, unorganized list of names by reading them one by one from top to bottom. If the name is at the bottom, it takes a long time.

* **Set ($O(1)$):** A set uses a **hash table**. It jumps straight to the item's memory address instantly, no matter how big the set is.

> **Rule of Thumb:** If you just need to check membership (`in`) frequently, always use a **set**.



# **Detailed Explanation :**    

### **The Hotel Mailbox Analogy**

Imagine a massive hotel with **10,000 mailboxes**.

If you want to find a guest's letter in a **List**, you have to walk down the hallway and check mailbox #1, then #2, then #3, all the way until you find it. If the letter isn't there, you wasted time checking all 10,000 boxes.

A **Hash Table (Set)** works completely differently. It uses a **secret mathematical formula (called a Hash Function)**.

---

#### **How It Works Step-by-Step**

1. **The Formula:** Whenever you put an item (like the number `30`) into a set, Python runs it through a math formula.
2. **The Direct Address:** The formula instantly spits out a specific spot number—let's say **Box #452**. Python drops `30` right into Box #452.
3. **The Instant Check:** When you ask Python `30 in my_set`, it doesn't search the set. It runs `30` through that exact same formula again. The formula instantly shouts: **"It's in Box #452!"**
4. **Jumping Straight There:** Python jumps *directly* to Box #452 without looking at any other box.

    Because it calculates the exact address using math instead of searching one by one, it takes the exact same amount of time whether the set has **5 items** or **5 million items**.

````python
# Memory Size Allocation of Data Structures in Computer Memory: List, Set, Tuple

# Creating Lists

import sys

an = []
print(sys.getsizeof(an))
print(an.__sizeof__())

an1 = [1]
print(sys.getsizeof(an1))
print(an1.__sizeof__())

an2 = [1,2]
print(sys.getsizeof(an2))
print(an2.__sizeof__())

# Creating Sets

bn = set()
print(sys.getsizeof(bn))
print(bn.__sizeof__())

bn1 = {1}
print(sys.getsizeof(bn1))
print(bn1.__sizeof__())

bn2 = {1,2}
print(sys.getsizeof(bn2))
print(bn2.__sizeof__())

# NOTE : As you can see, set data structure requires higher capacity of data storage in the computer's primary memory RAM

# Creating Tuple

cn = ()
print(sys.getsizeof(cn))
print(cn.__sizeof__())

cn1 = (1)
print(sys.getsizeof(cn1))
print(cn1.__sizeof__())

cn2 = (1,2)
print(sys.getsizeof(cn2))
print(cn2.__sizeof__())

# NOTE : As you can see, tuple data structure requires very little storage in the computer's primary memory of RAM.
````

### **Function Differences**

* **`.__sizeof__()`**: Measures only the raw core data of the object.
* **`sys.getsizeof()`**: Measures the core data **plus** Python's extra technical details, making the final number larger.

### **Why Tuples Use Less Memory**

* **Tuples** are fixed (immutable), so Python allocates the exact memory needed and nothing extra.
* **Lists and Sets** use more memory because Python pre-allocates extra hidden space behind the scenes so they can grow dynamically.

````python
# Normal Number Reassignment : ---------------------------

x = 5
y = x
x = 10
print(x,y)

### What happened? :

# 1. At first, python stores 5 to a computer memory location and label it as variable x.
# 2. Then python creates an another computer memory location and label it as variable y.
# 3. After that, python removes the number 5 from the computer memory location and assigns it 10 to that memory location
# Result : We have used 2 memory locations : x and y  where x stores 10 and y stores 5.

# List Reassignment : ------------------------------------

x1 = [1,2,3]
y1 = x1
x1[0] = 100
print(x1)
print(y1)

### What happened? :

# 1. At first, python stores the list [1,2,3] to a computer memory location and then label it as x1.
# 2. Then python creates another label for the exact same location where the list [1,2,3] is stored and label it as y1.
# 3. After that, python removes the 0th positional element from the list and replaces with 100.
# Result : We have used 1 memory location and label it as both : x1 and y1 where that memory location stores the list which is : [1,2,3]
````

````python
# So how can we assign the same list to different memory locations?

l11 = [1,2,3]                    # List 1
l12 = list(l11)                  # List 2
l13 = l11.copy()                 # List 3
l14 = l11[:]                     # List 4
l15 = l11

l12[0] = 100
l13[1] = 200
l14[2] = 300

print(l11,l12,l13,l14)

## Checking whether we created 4 distinct locations or not?

print(l11 is l12)
print(l11 is l13)
print(l11 is l14)
print(l11 is l15)

# Result : Successfully created 4 distinct list, and stores it in 4 distinct memory locations.
````

### **Set Equality, Cardinality, and Supersets**

#### **1. Same Cardinality (Size) but Different Elements**
If $|A| = |B|$ (same size) but the elements are different, neither is a subset or superset of the other.
* Example: $A = \{1, 2, 3\}$, $B = \{3, 4, 5\}$. Both have size 3, but $A \not\supseteq B$ and $B \not\supseteq A$.

#### **2. Same Cardinality and Identical Elements (Equal Sets)**
If $A = B$, they have the same size and identical elements. In mathematics, **every set is a superset (and subset) of itself**.
* Example: $A = \{1, 2, 3\}$, $B = \{1, 2, 3\}$. Here, $A \supseteq B$ and $B \supseteq A$ are both **True**.

#### **3. Proper Superset vs. Superset**
* **Superset ($\_ \ge \_$)**: $A$ contains all elements of $B$. They *can* be equal in size.
* **Proper Superset ($\_ > \_$)**: $A$ contains all elements of $B$ **and** $A$ is strictly larger than $B$; ($|A| > |B|$).

### **1. What is a Dictionary?**

After lists (which use square brackets `[]`) and tuples (which use round brackets `()`), Python introduces **dictionaries**, which use **curly brackets `{}**`.

Unlike lists or tuples where items are stored in a simple sequence, dictionary is a data structure where elements are written in **key-value pairs** separated by a colon (`:`):

1. **Left-hand side of the colon:** The **Key** (used to look up information).

2. **Right-hand side of the colon:** The **Value** (the actual data stored).

3. **Uniqueness:** Every key in a dictionary must be **unique** because you use it to access values. However, **values can be duplicated**.

### **2. What Can Be Used as a Dictionary Key?**

Not everything can be a key. The Golden Rule is that a key must be **immutable and hashable** (unchangeable).

* **Yes (Allowed):** Integers, floats, Booleans, and strings.

* **No (Not Allowed):** Lists, sets and other dictionaries because they are **mutable** (can be changed after creation).

     #### **What about Tuples? (Yes and No):**

* **Yes**, because tuples are generally immutable.

* **No**, if the tuple contains a mutable item inside it (like a list).

* **Conclusion:** Only tuples that contain strictly immutable items are allowed to be keys because they are **hashable (unchangable)**. and if they are hashable, then we can access a value easily.

### **3. What Can Be Used as a Dictionary Value?**

* **Anything!** There are **no restrictions** on dictionary values.
* Values can be integers, floats, Booleans, strings, lists, sets, tuples, or even other dictionaries.

### **4. Mutability, Copying, and Functions**

* Dictionaries are **mutable** (changeable), meaning all the properties of mutability that apply to lists also apply to dictionaries.

* **Copying:** If you want to copy a dictionary, you must explicitly use the **`copy()` method**, otherwise you will just create a reference to the original dictionary.

* **Functions:** When you pass a dictionary as an argument to a function, it is passed by reference (changes made inside the function affect the original dictionary).

#####  **UNDERSTANDING PASS-BY REFERENCE IN PYTHON** :

  * **What is an "argument"?**

      An argument is just the variable you put inside the parentheses when you run a function.

      Example: `my_function(my_dict)`

      Here, `my_dict` is the argument. It’s just normal data you are feeding into the function.

  * **What is a "default argument"?**

      A default argument is a backup option you set when creating the function, used only if someone forgets to give input.

      Example: `def my_function(my_dict = {})`
      (That's not what the lecture was talking about).

  * **What does "passed by reference" mean?**

      It just means Python **does not create a duplicate** of your dictionary when you pass it to a function.
      Instead, the function works directly on your **original dictionary**. So if the function adds or deletes something inside that dictionary, your original dictionary gets changed too. That’s all it means!

### **6. The Three Most Important Dictionary Methods**

Python provides built-in methods to work with dictionaries easily:

1. **`d.keys()`**: Returns a collection of all the keys in the dictionary.

2. **`d.values()`**: Returns a collection of all the values in the dictionary.

3. **`d.items()`**: Returns a list-like collection where every element is a **tuple** containing two items: the **key** and its corresponding **value** (an example of how Python uses tuples internally).

````python
# Finding the no. of occurence of each element within the dictionary :

d1 = {}                # Empty Dictionary

for i in l:
  d1[i] = d[i] + 1
print(d1)
````

````python
# Finding that element which has the highest number of occurences within the dictionary :

max_element = ""
max = 0

for i in d1:
  if d1[i] > max:
    max = d1[i]
    max_element = i
print(max_element,":", max)
````

````python
### Part 1: Without `.copy()` (They share the same data)

# Original dictionary
d1 = {"name": "Alice", "age": 25}

# Point d2 to d1 without copying
d2 = d1

# Modify d2 by adding a new key
d2["city"] = "New York"

print("d1:", d1)
print("d2:", d2)

# *Notice that `d1` changed too, even though we only edited `d2`!*
````

````python
### Part 2: With `.copy()` (They are completely independent)

# Original dictionary
d1 = {"name": "Alice", "age": 25}

# Create a true separate copy using .copy()
d2 = d1.copy()

# Modify d2
d2["city"] = "New York"

print("d1:", d1)
print("d2:", d2)

# This time, `d1` stays completely safe and unchanged because `d2` is a separate copy.*
````

## **Multiset**

````python
from collections import Counter

# Creating List

l = [1,2,3,4,5,5,5,5]
print(l)

# Conversion : List To Multiset

multiset = Counter(l)
print(multiset)

# Finding Frequencies of Element From MultiSet

print(multiset[5])
print(multiset[2])

# Adding Elements TO Multiset

multiset.update([2,3,4,5,8,9])
print(multiset)

# Subtracting Elements From Multiset To Decrease Frequency

## Example_1

multiset.subtract([1,2,3,4])
print(multiset)

## Example_2

multiset.subtract([1,2,3,4,5,6,7,8,9])
print(multiset)
````
