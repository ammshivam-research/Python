### Beginner-Level Python Operator Precedence

When evaluating an expression, Python follows this simplified order of priority (from **highest** to **lowest**). Operators on the same level are evaluated from **left to right**.

| Priority | Group / Operator | Symbol | Description |
| :---: | :--- | :---: | :--- |
| **1** | Parentheses | `()` | Forces evaluation order (always executed first) |
| **2** | Exponentiation | `**` | Power / Exponent |
| **3** | Multiplication & Division | `*`, `/`, `//`, `%` | Multiply, Divide, Floor Divide, Modulo (Remainder) |
| **4** | Addition & Subtraction | `+`, `-` | Plus and Minus |
| **5** | Comparisons | `>`, `<`, `>=`, `<=`, `==`, `!=` | Relational checks (returns `True` or `False`) |
| **6** | Logical NOT | `not` | Negates a boolean value |
| **7** | Logical AND | `and` | Returns `True` only if both are true |
| **8** | Logical OR | `or` | Returns `True` if at least one is true |

### Detailed Comparison: Division (`/`) vs. Floor Division (`//`)

| Feature | Standard Division (`/`) | Floor Division (`//`) |
| :--- | :--- | :--- |
| **Primary Function** | Performs standard division. | Performs division and rounds **down** to the next smallest integer (the "floor"). |
| **Result Type (Ints)** | **Always** returns a `float`. <br> *Example:* `6 / 3` $\rightarrow$ `2.0` | Returns an `int`. <br> *Example:* `6 // 3` $\rightarrow$ `2` |
| **Result Type (Floats)** | **Always** returns a `float`. <br> *Example:* `6.0 / 3` $\rightarrow$ `2.0` | Returns a `float` (representing a whole number). <br> *Example:* `6.0 // 3` $\rightarrow$ `2.0` |
| **Negative Numbers** | Preserves precision and sign. <br> *Example:* `-5 / 2` $\rightarrow$ `-2.5` | Rounds **down** (away from zero). <br> *Example:* `-5 // 2` $\rightarrow$ `-3` |

---

### Quick Rule of Thumb:
* Use `/` when you need **exact fractional parts** (e.g., calculating averages).
* Use `//` when you need **whole units** and want to discard any remainder (e.g., finding the number of full weeks in a given number of days).
