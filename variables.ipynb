# **Variables in Python**

A **variable** is a container used to store data values in computer's physical memory **RAM(Random Access Memory)**. In Python, you do not need to declare the type of variable explicitly unlike SQL; it is created the moment you assign a value to it using the `=` operator.

---

### Variable Naming Rules

| Rule | Allowed / Valid | Not Allowed / Invalid | Reason |
| :--- | :--- | :--- | :--- |
| **Starting Character** | `score`, `_temperature` | `2nd_place` | Must start with a letter or an underscore (`_`), never a number. |
| **Allowed Characters** | `user_score_1` | `user-score`, `user score` | Only letters, digits (`0-9`), and underscores (`_`) are allowed. |
| **Case Sensitivity** | `age`, `Age`, `AGE` | N/A | These are treated as 3 completely different variables. |
| **Reserved Words** | `class_name`, `my_def` | `class`, `def`, `if`, `for` | Reserved Python keywords cannot be used as variable names. |

# **Dynamic Typing in Python**

In Python, a variable does not have a fixed data type; instead, its type is determined entirely by its current value residing in the computer's physical memory **RAM (Random Access Memory)**. Variable names act as references or pointers, allowing a variable to pass through multiple different types across its lifetime via successive reassignments.

---

### Key Mechanics of Dynamic Typing & Operators

| Concept | Behavior in Memory / RAM | Details |
| :--- | :--- | :--- |
| **Dynamic Reassignment** | Types are determined by the current value and can change freely. | A variable can transition through multiple types (e.g., from `int` to `bool` to `float`) over its lifetime. |
| **True Division (`/`)** | Always allocates a float object in memory. | Performs true division and returns a float even when dividing by `1` or when the result is a whole number (e.g., `10 / 2` -> `5.0`). |
| **Floor Division (`//`)** | Preserves the integer type in memory. | Unlike `/`, floor division retains the `int` type by flooring the output (e.g., `10 // 2` -> `5`). |
| **Variable Independence** | Independent references to objects. | Reassigning one variable does not affect another variable that copied its earlier value. |

````python
### Code Examples

# 1. Successive Reassignment (Passing through multiple types, int to bool)
x = 42                  # Initialized as an integer object
print(type(x), x)

x = True                # Reassigned from int to bool
print(type(x), x)

x = 3.14                # Reassigned to float
print(type(x), x)

# 2. Division Operators Behavior (/ vs //)
result_true = 10 / 1    # / always returns float, even when dividing by 1 -> 10.0
result_floor = 10 // 1  # // preserves int type -> 10

print(type(result_true), result_true)
print(type(result_floor), result_floor)

# 3. Variable Independence & Reassignment
original_var = 50
copied_var = original_var   # copied_var references the same memory object

original_var = 100          # Reassigning original_var shifts its pointer
print(copied_var)           # Output: 50 (unaffected by original_var's change)
````
