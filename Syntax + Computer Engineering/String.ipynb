# **Python Quotes & Escape Characters text**

| Full Official Name              | Syntax / Representation  | Output Appearance           | Main Purpose                                                                       |
| :------------------------------ | :----------------------- | :-------------------------- | :--------------------------------------------------------------------------------- |
| **Single-Quoted String**        | `'Text'`                 | Text                        | Defines a standard string literal.                                                 |
| **Double-Quoted String**        | `"Text"`                 | Text                        | Defines a standard string literal, allowing single quotes inside without escaping. |
| **Multiline String Literal**    | `'''Text'''` or `"""Text"""` | Spans multiple lines     | Defines multiline strings or documentation docstrings.                             |
| **Newline Escape Character**    | `\n`                     | Moves text to the next line | Inserts a line break inside a string.                                              |
| **Horizontal Tab Escape Character** | `\t`                  | Adds a horizontal tab space | Aligns text with horizontal spacing.                                               |
| **Escaped Backslash Character** | `\\`                     | \                           | Displays a literal backslash character within a string.                            |
| **Escaped Single Quote Character** | `\'`                  | '                           | Embeds a literal single quote inside a single-quoted string.                       |
| **Escaped Double Quote Character** | `\"`                  | "                           | Embeds a literal double quote inside a double-quoted string.                       |

````python
# 1. Single Quotes
text_single = 'Hello, World!'
print(text_single)

# 2. Double Quotes
text_double = "Hello, World!"
print(text_double)

# 3. Triple Quotes (Multiline)
text_triple = """This is a
multiline string."""
print(text_triple)

# 4. Newline Escape Character (\n)
print("Line 1\nLine 2")

# 5. Tab Escape Character (\t)
print("Column1\tColumn2")

# 6. Backslash Escape Character (\\)
print("C:\\Users\\Name")

# 7. Escaped Single Quote (\')
print('It\'s a wonderful day')

# 8. Escaped Double Quote (\")
print("He said, \"Hello!\"")
````

# **3 Quick Extra Concepts**
1. **Why Triple Quotes Act as Comments?**

    If you write text inside """ ... """ without saving it to a variable, Python simply ignores it. That is why programmers use it to write multi-line comments.

2. **Why Two Backslashes (\\) Are Needed?**

    A single \ tells Python "a special code is coming next" (like \n). If you want to print an actual backslash symbol, you type \\ so Python knows you mean the literal character \.

3. **Why Escaped Quotes (\' and \") Exist?**

    Quotes mark where a string starts and ends. If you want to put a quote inside a string using the same quote type, \' or \" tells Python "this is just normal text, do not close the string yet."

# **Python String Methods Reference**

| Title Name   | Full Official Name         | Code with String                    | Code with Variable                          | Output Looks Like | Main Purpose                                          |
| :----------- | :------------------------- | :---------------------------------- | :------------------------------------------ | :---------------- | :---------------------------------------------------- |
| `lower()`    | **Lower Method**           | `"HELLO".lower()`                   | `text = "HELLO"; text.lower()`              | `'hello'`         | Converts all letters to lowercase.                    |
| `upper()`    | **Upper Method**           | `"hello".upper()`                   | `text = "hello"; text.upper()`              | `'HELLO'`         | Converts all letters to uppercase.                    |
| `capitalize()` | **Capitalize Method**    | `"hello world".capitalize()`        | `text = "hello world"; text.capitalize()`   | `'Hello world'`   | Capitalizes only the very first letter of the string. |

| `isalnum()`  | **Is Alphanumeric Check**  | `"Python3".isalnum()`               | `text = "Python3"; text.isalnum()`          | `True`            | Checks if characters are only letters or numbers.     |
| `strip()`    | **Strip Method**           | `" hi ".strip()`                    | `text = " hi "; text.strip()`               | `'hi'`            | Removes whitespace from both ends.                    |
| `lstrip()`   | **Left Strip Method**      | `" hi ".lstrip()`                   | `text = " hi "; text.lstrip()`              | `'hi '`           | Removes whitespace only from the left side.           |

````python
# 1. lower() - Converts all letters to lowercase
print("HELLO".lower())

# 2. upper() - Converts all letters to uppercase
print("hello".upper())

# 3. capitalize() - Capitalizes the first letter of the string
print("hello world".capitalize())

# 4. title() - Capitalizes the first letter of every word
print("hello world".title())

# 5. swapcase() - Swaps case (upper to lower, lower to upper)
print("HeLLo".swapcase())

# 6. islower() - Checks if all letters are lowercase
print("hello".islower())

# 7. isupper() - Checks if all letters are uppercase
print("HELLO".isupper())

# 8. istitle() - Checks if every word starts with a capital letter
print("Hello World".istitle())

# 9. isdigit() - Checks if all characters are numbers
print("12345".isdigit())

# 10. isalpha() - Checks if all characters are alphabetic letters
print("Python".isalpha())

# 11. isalnum() - Checks if characters are only letters or numbers
print("Python3".isalnum())

# 12. strip() - Removes whitespace from both ends
print("  hi  ".strip())

# 13. lstrip() - Removes whitespace from the left side
print("  hi  ".lstrip())

# 14. rstrip() - Removes whitespace from the right side
print("  hi  ".rstrip())

# 15. startswith() - Checks if the string starts with specific text
print("hello".startswith("he"))

# 16. endswith() - Checks if the string ends with specific text
print("hello".endswith("lo"))

# 17. count() - Counts how many times a substring appears
print("banana".count("a"))

# 18. index() - Finds the first position (index) of a character
print("hello".index("e"))

# 19. replace() - Replaces parts of a string with new text
print("hello".replace("l", "z"))
````

# **Important Points**

# **rfind()**

**Search from the right:** Instead of starting from the beginning (left), rfind() starts searching from the very end (right) of the text.

**Safe -1:** If it cannot find what you are looking for, it quietly returns -1 instead of crashing your program.

````python
text = "banana"

print(text.rfind("a"))  # Output: 5 (Finds the last 'a' at the back)
print(text.rfind("z"))  # Output: -1 (Safe fallback, doesn't crash)
````

# **strip() with Custom Characters**

When you give strip() a list of characters (like ".+"), Python treats them like individual items in a junk pile.

.strip() only looks at the outer edges (the very beginning and the very end) of the string.

Once it hits a character that is not in the given set ('yxln')—like 'L' on the left and 'o' on the right—it stops stripping immediately. It never scans or modifies anything in the middle of the string.

````python
text = "...hello++"

# Strips any dots or plus signs sitting on the outer edges
print(text.strip(".+"))  # Output: 'hello'
````

# **startswith() and endswith() Are Strict (Case-Sensitive)**

Python cares a lot about capital and small letters. If your text starts with a big letter ('H'), asking Python if it starts with a small letter ('h') will result in a False

````python
text = "Hello"

print(text.startswith("He"))  # Output: True
print(text.startswith("he"))  # Output: False (Big 'H' vs small 'h')
````

# **count() Does Not Overlap**

When count() finds a match, it moves forward and doesn't look backward. It never steals or reuses letters.

````python
# Looking for "aa" inside "aaaa":
# It checks the first two letters, then jumps to the next two. Total = 2.
print("aaaa".count("aa"))  # Output: 2
````

# **replace() with a Limit**

Normally, replace() changes all matches it finds. If you add a number at the end (like 2), you are telling Python: "Stop changing things after you've done it 2 times!"

````python
text = "ha ha ha ha"

# Change only the first 2 "ha"s into "ho"
print(text.replace("ha", "ho", 2))  # Output: 'ho ho ha ha'
````

# **index() (Finding Position & Crashing if Missing)**

index() tells you the exact position number where a character first appears.

**Watch out:** Unlike rfind(), if it can't find what you asked for, it crashes your code with an error instead of returning -1

````python
text = "hello"

print(text.index("e"))  # Output: 1 (The letter 'e' is at position 1)
````

# **Caesar Cipher Python Code**

**What is Caesar Cipher?**  
It is a very old and simple way to hide a message.  
You just move every letter in the alphabet by a fixed number of steps.


**Who invented it?**  
Julius Caesar (a famous Roman leader) used it long ago to send secret messages to his soldiers.


**How does it work?**  
1. Choose a secret number (called the **key** or **shift**).  
   Example: key = 3  
2. Take each letter of your message.  
3. Move that letter forward in the alphabet by the key number.  
4. If you go past Z, start again from A.

**Example** (key = 3)

| Original letter | A | B | C | D | E | F | G | H | I | J | K | L | M |
|-----------------|---|---|---|---|---|---|---|---|---|---|---|---|---|
| Secret letter   | D | E | F | G | H | I | J | K | L | M | N | O | P |

| Original letter | N | O | P | Q | R | S | T | U | V | W | X | Y | Z |
|-----------------|---|---|---|---|---|---|---|---|---|---|---|---|---|
| Secret letter   | Q | R | S | T | U | V | W | X | Y | Z | A | B | C |

So the word **HELLO** becomes **KHOOR**.


**How to read the secret message?**  
Just move the letters **backward** by the same key number.


**Important points for beginners**
- Very easy to understand and use.
- Only 25 possible keys (because alphabet has 26 letters).
- Not safe for real secrets today — anyone can try all keys quickly.
- Great for learning the basic idea of encryption.




````python
#Caesar Cipher Python Code (Beginner Level)

alpha='abcdefghijklmnopqrstuvwxyz'
s='chennai'
#I expect to output fkhqqdl
t=''
i=0
t+=alpha[((alpha.index(s[i])+3)%26)]
t+=alpha[((alpha.index(s[i+1])+3)%26)]
t+=alpha[((alpha.index(s[i+2])+3)%26)]
t+=alpha[((alpha.index(s[i+3])+3)%26)]
t+=alpha[((alpha.index(s[i+4])+3)%26)]
t+=alpha[((alpha.index(s[i+5])+3)%26)]
t+=alpha[((alpha.index(s[i+6])+3)%26)]
print(t)
````

**The error IndexError:**

String index out of range occurs because you are trying to access an index that is outside the bounds of the alphabet string.

**Why it happens:**

The alphabet string alpha has 26 letters (indices 0 to 25).
The letter 'y' is at index 24 in alpha.
When you add the key k = 3, you calculate 24 + 3 = 27.
Since the highest index in alpha is 25, trying to access alpha[27] crashes your program.

**How to fix it:**

To handle shifts that go past 'z', you must use the modulo operator (% 26) to wrap around back to the beginning of the alphabet (e.g., 27 % 26 becomes index 1, which is 'b').
