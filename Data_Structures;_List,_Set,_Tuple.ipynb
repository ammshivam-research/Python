## **2D List (Matrix)**

Think of 2D data like a grid or a spreadsheet with rows and columns.

A normal list like [1, 2, 3] is just a single straight line (1D). But when you put multiple lists inside one big list, you get a grid of rows and columns (2D).

To grab any number, you just need two coordinates: the row number and the column number.

#### **Visual Table (with Row and Column Indexes)**

|  | Column 0 | Column 1 | Column 2 |
| --- | --- | --- | --- |
| **Row 0** | `1` | `2` | `3` |
| **Row 1** | `4` | `5` | `6` |
| **Row 2** | `7` | `8` | `9` |


#### **How to Access Elements**

To grab any item, use two brackets: `grid[row_index][column_index]`.

```python
grid = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Get the number at Row 0, Column 2
print(grid[0][2])  # Output: 3

# Get the number at Row 1, Column 1 (the middle one)
print(grid[1][1])  # Output: 5

# Get the number at Row 2, Column 0
print(grid[2][0])  # Output: 7

```

### **Accessing (checking if an item exists using in) is much faster in either set or a list?**

Accessing (checking if an item exists using `in`) is **much faster in a set** than in a list.

### **Why Sets are Faster**

* **List ($O(n)$):** A list checks items **one by one** from start to finish. If the item is at the end or not there at all, it has to look through every single element.

    Like looking for a name in a long, unorganized list of names by reading them one by one from top to bottom. If the name is at the bottom, it takes a long time.

* **Set ($O(1)$):** A set uses a **hash table**. It jumps straight to the item's memory address instantly, no matter how big the set is.

> **Rule of Thumb:** If you just need to check membership (`in`) frequently, always use a **set**.



# **Detailed Explanation :**    

### **The Hotel Mailbox Analogy**

Imagine a massive hotel with **10,000 mailboxes**.

If you want to find a guest's letter in a **List**, you have to walk down the hallway and check mailbox #1, then #2, then #3, all the way until you find it. If the letter isn't there, you wasted time checking all 10,000 boxes.

A **Hash Table (Set)** works completely differently. It uses a **secret mathematical formula (called a Hash Function)**.

---

#### **How It Works Step-by-Step**

1. **The Formula:** Whenever you put an item (like the number `30`) into a set, Python runs it through a math formula.
2. **The Direct Address:** The formula instantly spits out a specific spot number—let's say **Box #452**. Python drops `30` right into Box #452.
3. **The Instant Check:** When you ask Python `30 in my_set`, it doesn't search the set. It runs `30` through that exact same formula again. The formula instantly shouts: **"It's in Box #452!"**
4. **Jumping Straight There:** Python jumps *directly* to Box #452 without looking at any other box.

    Because it calculates the exact address using math instead of searching one by one, it takes the exact same amount of time whether the set has **5 items** or **5 million items**.

````python
# Memory Size Allocation of Data Structures in Computer Memory: List, Set, Tuple

# Creating Lists

an = []
print(sys.getsizeof(an))
print(an.__sizeof__())

an1 = [1]
print(sys.getsizeof(an1))
print(an1.__sizeof__())

an2 = [1,2]
print(sys.getsizeof(an2))
print(an2.__sizeof__())

# Creating Sets

bn = {}
print(sys.getsizeof(bn))
print(bn.__sizeof__())

bn1 = {1}
print(sys.getsizeof(bn1))
print(bn1.__sizeof__())

bn2 = {1,2}
print(sys.getsizeof(bn2))
print(bn2.__sizeof__())

# Creating Tuple

cn = ()
print(sys.getsizeof(cn))
print(cn.__sizeof__())

cn1 = (1)
print(sys.getsizeof(cn1))
print(cn1.__sizeof__())

cn2 = (1,2)
print(sys.getsizeof(cn2))
print(cn2.__sizeof__())

# NOTE : As you can see, tuple data structure requires very little storage in the computer memory.
````

### **Function Differences**

* **`.__sizeof__()`**: Measures only the raw core data of the object.
* **`sys.getsizeof()`**: Measures the core data **plus** Python's extra technical details, making the final number larger.

### **Why Tuples Use Less Memory**

* **Tuples** are fixed (immutable), so Python allocates the exact memory needed and nothing extra.
* **Lists and Sets** use more memory because Python pre-allocates extra hidden space behind the scenes so they can grow dynamically.

````python
# Normal Number Reassignment : ---------------------------

x = 5
y = x
x = 10
print(x,y)

### What happened? :

# 1. At first, python stores 5 to a computer memory location and label it as variable x.
# 2. Then python creates an another computer memory location and label it as variable y.
# 3. After that, python removes the number 5 from the computer memory location and assigns it 10 to that memory location
# Result : We have used 2 memory locations : x and y  where x stores 10 and y stores 5.

# List Reassignment : ------------------------------------

x1 = [1,2,3]
y1 = x1
x1[0] = 100
print(x1)
print(y1)

### What happened? :

# 1. At first, python stores the list [1,2,3] to a computer memory location and then label it as x1.
# 2. Then python creates another label for the exact same location where the list [1,2,3] is stored and label it as y1.
# 3. After that, python removes the 0th positional element from the list and replaces with 100.
# Result : We have used 1 memory location and label it as both : x1 and y1 where that memory location stores the list which is : [1,2,3]
````

````python
# So how can we assign the same list to different memory locations?

l11 = [1,2,3]                    # List 1
l12 = list(l11)                  # List 2
l13 = l11.copy()                 # List 3
l14 = l11[:]                     # List 4
l15 = l11

l12[0] = 100
l13[1] = 200
l14[2] = 300

print(l11,l12,l13,l14)

## Checking whether we created 4 distinct locations or not?

print(l11 is l12)
print(l11 is l13)
print(l11 is l14)
print(l11 is l15)

# Result : Successfully created 4 distinct list, and stores it in 4 distinct memory locations.
````

### **LIST FUNCTIONS**

````python
# List Functions :

sd = [10,2,38,98,7,0,6]

# Want to add an element in a list?
sd.append(8)
print(sd)

# Want to add an element in a list at a specific position?
sd.insert(2,9)
print(sd)

# Want to arrange the elements in ascending order to change their positioning in a list?
sd.sort()
print(sd)

# Want to arrange the elements in descending order to change their positioning in a list?
sd.sort(reverse = True)
print(sd)

# Want to reverse the elements to change their positioning in a list?
sd.reverse()
print(sd)
````

````python
# Demonstration of .remove() vs .pop()

# 1. Using .remove()
numbers = [10, 20, 30, 20, 40]
print("Original list:", numbers)

numbers.remove(20) # Removes the first '20'
print("After remove(20):", numbers)

# 2. Using .pop()
popped_val = numbers.pop(2) # Removes and returns the element at index 2 (which is 20)
print("Popped value:", popped_val)
print("After pop(2):", numbers)

# 3. Using .pop() with no index (removes last element)
last_val = numbers.pop()
print("Popped last value:", last_val)
print("After pop():", numbers)
````
