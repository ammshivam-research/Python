# **Immutable Object**

1. String
2. Integer
3. Float
4. Boolean
5. Tuple

## **Indexing and Slicing Compatibility**

| Immutable Data Type | Indexing | Slicing |
|---|---|---|
| String (`str`) | ✅ Supported | ✅ Supported |
| Integer (`int`) | ❌ Not Supported | ❌ Not Supported |
| Float (`float`) | ❌ Not Supported | ❌ Not Supported |
| Tuple (`tuple`) | ✅ Supported | ✅ Supported |
| Boolean (`bool`) | ❌ Not Supported | ❌ Not Supported |

### **String**

```python
name = "PYTHON"

# Indexing
name[0]       # 'P'
name[-1]      # 'N'

# Slicing
name[0:3]     # 'PYT'
name[::-1]    # 'NOHTYP'
```

### **Integer (int)**

```python
number = 12345

number[0]     # TypeError
number[0:3]   # TypeError
```

### **Float (float)**

```python
number = 123.45

number[0]     # TypeError
number[0:3]   # TypeError
```

### **Tuple (tuple)**

```python
numbers = (10, 20, 30, 40, 50)

# Indexing
numbers[0]       # 10
numbers[-1]      # 50

# Slicing
numbers[0:3]     # (10, 20, 30)
numbers[::-1]    # (50, 40, 30, 20, 10)
```

### **Boolean (bool)**

```python
value = True

value[0]      # TypeError
value[0:1]    # TypeError
```
