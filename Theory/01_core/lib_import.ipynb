**Question 1** : What is a module?


A module is a single Python file (.py) containing reusable code, such as functions, classes, and variables.

**1. Core Anatomy**

*   **File Format** : Saved with a .py extension.
*   **Contents** : Functions, global variables, and classes.
*   **Primary Goal** : Code reusability and clean organization.

| Type                   | Description                          | Example                      |
| :--------------------- | :----------------------------------- | :--------------------------- |
| **Built-in Modules**   | Pre-installed with standard Python.  | `math`, `calendar`, `random` |
| **User-Defined Modules** | Custom Python files created by you. | `my_utils.py`, `database.py` |
| **Third-Party Modules**| External packages installed via pip. | `pandas`, `numpy`, `requests`|

**Question 2** : Explain different types of import statements in Python

### 1. `import calendar`

* **What it does:** Imports the whole module.
* **How to use:** You must prefix functions with the module name.
* **Example:** `calendar.month(2026, 10)`

### 2. `from calendar import *`

* **What it does:** Imports **everything** inside the module directly into your script.
* **How to use:** Call functions directly without any prefix.
* **Example:** `month(2026, 10)`

### 3. `from calendar import month`

* **What it does:** Imports **only the `month` function** from the module.
* **How to use:** Call `month` directly; other functions from `calendar` remain unavailable unless imported.
* **Example:** `month(2026, 10)`

### 4. `from calendar import month, calendar`

* **What it does:** Imports **multiple specific items** (`month` and `calendar`) directly.
* **How to use:** Call both functions directly without prefixes.
* **Example:** `calendar(2026)`

### 5. `from calendar import month as m`

* **What it does:** Imports the `month` function and **renames it locally** to `m` (using an alias).
* **How to use:** Call the function using its new short name.
* **Example:** `m(2026, 10)`

````python
import calendar

# You must write the module name before the function
print(calendar.month(2026, 10))
````

````python
from calendar import *

# You can call the function directly without any prefix
print(month(2026, 10))
````

````python
from calendar import month

# Only the 'month' function is available to use directly
print(month(2026, 10))
````

````python
from calendar import month, calendar as calendar_year

# Multiple items are imported and ready to use directly
# We use 'calendar_year' (the calendar function) instead of 'year'
print(calendar_year(2026))
````

````python
from calendar import month as m

# The function is renamed locally to 'm' for convenience
print(m(2026, 10))
````

# 1. ```math``` **Module**

Speed: It runs on C-language under the hood, making calculations super fast.

1.   **What is it?**

       Think of math as a scientific calculator built right into Python. It handles advanced math operations that go beyond basic addition, subtraction, or multiplication.
2.   **Key Concepts**

       **Constants**: Fixed mathematical values that never change, like π (Pi).

       **Speed**: It runs on C-language under the hood, making calculations super fast.



````python
## All Important Syntaxes & Examples

import math

# 1. Square Root
result1 = math.sqrt(16)  # Output: 4.0
print(result1)

# 2. Power (x raised to y)
result2 = math.pow(2, 3)  # Output: 8.0
print(result2)

# 3. Rounding Up (Ceil)
result3 = math.ceil(4.2)  # Output: 5
print(result3)

# 4. Rounding Down (Floor)
result4 = math.floor(4.8)  # Output: 4
print(result4)

# 5. Absolute Value (Removes negative sign)
result5 = math.fabs(-7.5)  # Output: 7.5
print(result5)

# 6. Factorial (5 * 4 * 3 * 2 * 1)
result6 = math.factorial(5)  # Output: 120
print(result6)

# 7. Constants
pi_val = math.pi  # Output: 3.141592653589793
print(pi_val)

e_val = math.e  # Output: 2.718281828459045
print(e_val)

# 8. Logarithms
log10_val = math.log10(100)  # Output: 2.0
print(log10_val)
````

# 2. ```random``` **Module**

1.   **What is it?**

        Think of random as a digital dice roller or coin flipper. It lets your program pick items, shuffle lists, or generate numbers unpredictably.

2.   **Key Concepts**

        **Pseudo-Random**: Computers can't make true random choices. They use formulas that simulate randomness, which is good enough for almost all daily programming tasks.

````python
## All Important Syntaxes & Examples

import random

# 1. Random Integer (between a and b, inclusive)
num = random.randint(1, 6)  # Output: Random integer like 1, 3, or 6
print(num)

# 2. Random Float (between 0.0 and 1.0)
val = random.random()  # Output: e.g., 0.4328...
print(val)

# 3. Random Pick from a List
colors = ["red", "blue", "green"]
pick = random.choice(colors)  # Output: e.g., "blue"
print(pick)

# 4. Pick Multiple Items (with replacement)
picks = random.choices(colors, k=2)  # Output: e.g., ['red', 'red']
print(picks)

# 5. Pick Unique Items (without replacement)
samples = random.sample(colors, k=2)  # Output: e.g., ['green', 'blue']
print(samples)

# 6. Shuffle a List (changes the original list directly)
numbers = [1, 2, 3, 4, 5]
print(random.shuffle(numbers)) # Output: numbers becomes something like [3, 1, 5, 2, 4]
````

### Why does `random.shuffle()` return `None`?

In Python, methods or functions that modify a list **in-place** (directly altering the original list) return `None` by design. This includes methods like `list.sort()`, `list.reverse()`, and `random.shuffle()`.

To view the shuffled list, you must first shuffle it, and then print the list variable on a separate line.

````python
# Correct way to shuffle and display a list:
numbers = [1, 2, 3, 4, 5]

# 1. Shuffle the list (this modifies 'numbers' directly)
random.shuffle(numbers)

# 2. Print the modified list
print(numbers)
````

# 3. **```datetime``` Module**


1.   **What is it?**

       Think of datetime as a calendar and clock combined. It helps Python understand dates, times, and time gaps, rather than just treating dates as plain text strings.
2.   **Key Concepts**

       **Date**: Year, month, day.

       **Time**: Hours, minutes, seconds, microseconds.

       **Timedelta**: Represents a duration (e.g., "difference of 5 days").

````python
## All Important Syntaxes & Examples

import datetime

# 1. Today's Date
today = datetime.date.today()  # Output: e.g., 2026-10-17
print(today)

# 2. Current Date and Time
now = datetime.datetime.now()  # Output: e.g., 2026-10-17 16:46:00.123456
print(now)

# 3. Creating a Specific Date (Year, Month, Day)
custom_date = datetime.date(2026, 12, 25)
print(custom_date)

# 4. Creating a Specific Date & Time
custom_dt = datetime.datetime(2026, 12, 25, 10, 30, 0)
print(custom_dt)

# 5. Date Arithmetic (Adding / Subtracting Days)
ten_days = datetime.timedelta(days=10)
future_date = today + ten_days  # Adds 10 days to today
print(future_date)

# 6. Finding Difference Between Two Dates
diff = custom_date - today
print(diff.days)  # Output: Number of days remaining

# 7. Format Date to String (strftime)
formatted = now.strftime("%Y-%m-%d %H:%M:%S")  # e.g., "2026-10-17 16:46:00"
print(formatted)

# 8. Parse String into Date (strptime)
date_obj = datetime.datetime.strptime("17/10/2026", "%d/%m/%Y")
print(date_obj)
````

**Question 1** : Simulate the sum of 3 fair dices

````python
import random

# Generate a random number from 1 to 6 for the first die (7 is exclusive)
rndm1 = random.randrange(1, 7)

# Generate a random number from 1 to 6 for the second die
rndm2 = random.randrange(1, 7)

# Generate a random number from 1 to 6 for the third die
rndm3 = random.randrange(1, 7)

# Calculate the total sum of all three dice rolls
sum = rndm1 + rndm2 + rndm3

# Display the final sum
print("The Sum of 3 random dices are : ", sum)
````
