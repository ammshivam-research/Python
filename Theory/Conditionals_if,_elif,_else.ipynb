# **Python Conditional Statements: `if`, `elif`, and `else`**



## 1. What are Conditional Statements?

Conditional statements are used to make decisions in a Python program.

They allow Python to execute different blocks of code depending on whether a given condition is `True` or `False`.

Python provides three main keywords for conditional statements:

- `if`
- `elif`
- `else`

## 2. Anatomy of a Basic `if` and `else` Statement

A basic conditional statement consists of the following components:

### A. `if` keyword

- Tells Python that you are about to check a condition.
- Python evaluates the condition to determine whether it is `True` or `False`.

### B. The condition (`age >= 18`)

- A condition is an expression that evaluates to either `True` or `False`.
- Python checks whether the given condition is satisfied.

### C. The colon (`:`)

- The colon is a mandatory symbol.
- It marks the end of the condition line and indicates the beginning of a code block.

### D. Indentation

- Indentation means adding spaces at the beginning of a line.
- It identifies which statements belong to a particular conditional block.
- The indented code inside the `if` block executes only when its condition is `True`.
- If the condition is `False`, Python skips that block and moves to the next part of the conditional statement.

### E. How `else` Works

- The `else` keyword does not have its own condition.
- It executes when the preceding `if` condition is `False`.
- If the `if` condition is `True`, Python executes the `if` block and skips the `else` block.
- If the `if` condition is `False`, Python executes the `else` block instead.

## 3. What is the `elif` Keyword?

`elif` stands for else if.

- It is used when we need to check multiple conditions in a program.
- It allows Python to check another condition if the preceding `if` or `elif` condition is `False`.
- You can use multiple `elif` statements in a single conditional chain.
- Python executes the first block whose condition evaluates to `True`.
- Once a condition is satisfied, Python skips all the remaining `elif` and `else` blocks.

### How does `elif` work?

Python follows this sequence:

1. First, Python checks the `if` condition.
2. If it is `True`, Python executes the `if` block and skips the remaining blocks.
3. If it is `False`, Python checks the first `elif` condition.
4. If that is also `False`, Python checks the next `elif` condition.
5. If all conditions are `False`, Python executes the `else` block, if provided.

Important rules:

1. An `if` statement must come first in a conditional chain.
2. You can use zero or more `elif` statements.
3. The `else` statement is optional.
4. An `else` statement must come after all `elif` statements.
5. Only the first block whose condition is `True` gets executed in an `if-elif-else` chain.

## 4. Difference Between Multiple `if` Statements and `if-elif`

| Multiple `if` statements                                    | `if-elif` statements                            |
| ----------------------------------------------------------- | ----------------------------------------------- |
| Each `if` condition is checked independently.               | Conditions are checked in sequence.             |
| Multiple blocks can execute if their conditions are `True`. | Only the first matching block executes.         |
| Each `if` is a separate conditional statement.              | All conditions belong to one conditional chain. |

## 5. Python Code Examples


```python
if condition1:
    # Executes if condition1 is True

elif condition2:
    # Executes if condition2 is True

elif condition3:
    # Executes if condition3 is True

else:
    # Executes if all conditions are False
```

```python
# Example 1: Basic if-else

age = 20

if age >= 18:
    print("Eligible")
else:
    print("Not eligible")
```

```python
# Example 2: Using if-elif-else

marks = 85

if marks >= 90:
    print("Grade A+")

elif marks >= 80:
    print("Grade A")

elif marks >= 70:
    print("Grade B")

elif marks >= 60:
    print("Grade C")

else:
    print("Fail")
```

```python
# Example 4: if-elif Statements

marks = 85

if marks >= 90:
    print("Grade A+")

elif marks >= 80:
    print("Grade A")

elif marks >= 70:
    print("Grade B")


## Why? The first condition is `False`, but the second condition is `True`. Python executes the second block and skips the remaining conditions.
```

# **Topic 1: Boundary Conditions in Comparisons**

Boundary conditions deal with the exact edge values where your logic flips from True to False.
Choosing the wrong operator (< vs <=) can accidentally exclude or include critical data points (like an exact age of 18 or a passing score of 40).

**Key Takeaway :**

Always check whether your business logic or rule needs to include the exact boundary value (>=, <=) or exclude it (>, <).

| Operator | Name                     | Mathematical Meaning | Edge Behavior at Exact Match             |
| :------- | :----------------------- | :------------------- | :--------------------------------------- |
| `<`      | **Strictly Less Than**   | $x < y$              | Excludes $y$ (False if value equals $y$) |
| `<=`     | **Less Than or Equal To**| $x \le y$            | Includes $y$ (True if value equals $y$)  |
| `>`      | **Strictly Greater Than**| $x > y$              | Excludes $y$ (False if value equals $y$) |
| `>=`     | **Greater Than or Equal To** | $x \ge y$        | Includes $y$ (True if value equals $y$)  |
| `==`     | **Equal To**             | $x = y$              | Matches exact equality                   |
| `!=`     | **Not Equal To**         | $x \neq y$           | True for anything except exact equality  |

```python
## Example : -------------------------

score = 40

# Case 1: Using strict inequality (Exclusive)
if score > 40:
    print("Passed")
else:
    print("Failed")  # Output: This runs because 40 is not > 40

# Case 2: Using inclusive inequality (Inclusive)
if score >= 40:
    print("Passed")  # Output: This runs because 40 is included in >= 40
else:
    print("Failed")
```

# **Topic 2: Code Placed Outside the if-else Block (Unindented)**

When you write code outside the if-else block (meaning it has no spaces or indentation at the start), Python runs it every single time, no matter which path (if or else) was taken.

| Where the code is     | Indentation                              | When does it run?                              |
| :-------------------- | :--------------------------------------- | :--------------------------------------------- |
| **Inside if**         | Indented (has spaces)                    | Only when the condition is True.               |
| **Inside else**       | Indented (has spaces)                    | Only when the condition is False.              |
| **Outside the block** | No indentation (starts at the very left) | Always runs, right after the if-else finishes. |

```python
is_authenticated = False

if is_authenticated:
    # SCENARIO A: If the condition is True, this runs (and the else is skipped)
    print("Access granted to dashboard.")
else:
    # SCENARIO B: If the condition is False, this runs instead
    print("Access denied. Please log in.")

# This line has no indentation, so it sits outside the if-else block.
# It runs EVERY SINGLE TIME, regardless of whether Scenario A or Scenario B happened.
print("Session audit log updated.")
```

# **Topic 3: Syntax Error from a Misplaced Statement**

Python links if and else together as a single team. If you accidentally stick a random line of code right in the middle between them, you break the connection. Python gets confused, thinks the else is completely orphaned without an if, and crashes with a SyntaxError.

**How to Fix It**:

Never put anything between the end of your if block and the start of your else statement. They must stick directly together!

```python
score = 45

if score >= 40:
    print("Passed")

# SCENARIO A (The Break): If you stick this unindented line right here...
print("Checking score status...")

# SCENARIO B (The Crash): Python now sees 'else' as completely alone
# because the line above broke the connection. It throws a SyntaxError!
else:
    print("Failed")
```

# **Topic 4 : Multiple Independent if Statements**

When you write multiple separate if statements one after another, Python treats them as completely independent questions. It will check every single if statement, even if the one before it was already true.

This is different from an if-elif chain, which stops checking the moment it finds a matching condition.

| Structure Type                | How Python Checks It                                         | Can multiple blocks run?                                   |
| :---------------------------- | :----------------------------------------------------------- | :--------------------------------------------------------- |
| **Independent if statements** | Checks every single `if` block, one by one.                  | Yes, multiple blocks can run if their conditions are true. |
| **if-elif-else chain**        | Stops checking as soon as it finds the first true condition. | No, only the first matching block runs.                    |

```python
## CODE EXAMPLE : -------------------------

score = 85

# --- INDEPENDENT IF STATEMENTS ---
# Python checks ALL of these rules separately.

if score >= 50:
    # SCENARIO A: This condition is True, so this runs.
    print("You passed the exam.")

if score >= 80:
    # SCENARIO B: Because this is a separate 'if', Python checks this too!
    # Since 85 >= 80 is also True, this runs as well.
    print("You got an A grade!")

if score == 100:
    # SCENARIO C: This condition is False, so Python skips it.
    print("Perfect score!")
```

# **Topic 5 : Complete Guide to Logical Operators (and, or, not)**

When you need to make decisions based on multiple rules, Python gives you three logical operators to combine or flip conditions: and, or, and not.

| Operator | What it Means                             | When is the overall condition True?                                               | Example Scenario                                                           |
| :------- | :---------------------------------------- | :-------------------------------------------------------------------------------- | :------------------------------------------------------------------------- |
| **and**  | Requires all conditions to be met         | Only if both sides are True. If even one side is False, the whole thing fails.    | `age >= 18 and has_id == True` (Needs both age and ID)                     |
| **or**   | Requires at least one condition to be met | If any one side is True (or both are True). It only fails if every side is False. | `is_student == True or has_coupon == True` (Gets discount if either is true) |
| **not**  | Flips the current boolean value           | If the original condition is False (it turns it True), or vice versa.             | `not is_banned` (Runs if user is not banned)                               |

```python
score = 85
attendance = 90
is_banned = False

# --- 1. USING 'and' ---

if score >= 50 and attendance >= 75:
    # SCENARIO A: Both parts are True (85 >= 50 and 90 >= 75), so this runs.
    print("Qualified with 'and'.")
else:
    # SCENARIO B: If either side fails, this runs.
    print("Failed 'and' check.")


# --- 2. USING 'or' ---
# Even though score >= 90 is False, attendance >= 75 is True.
# Because it's an 'or', having at least ONE True is enough!

if score >= 90 or attendance >= 75:
    # SCENARIO A: At least one side is True, so this runs.
    print("Passed via 'or' condition.")
else:
    # SCENARIO B: Only runs if BOTH sides are False.
    print("Failed 'or' check.")


# --- 3. USING 'not' ---

if not is_banned:
    # SCENARIO A: 'is_banned' is False, but 'not False' turns it into True!
    # Therefore, this runs.
    print("Access allowed: User is clean.")
else:
    # SCENARIO B: Runs only if 'is_banned' was actually True.
    print("Access blocked: User is banned.")
```

# **Topic 6 : Truthy and Falsy Values (Using Plain Values as Conditions)**

In Python, an if statement doesn't always need a comparison like `x > 0`. You can put a plain variable directly into the condition.

Python automatically figures out if it counts as `True` (Truthy) or `False` (Falsy).

| Category | Meaning to Python | What it looks like in code                                      | What the if statement does                          |
| :------- | :---------------- | :-------------------------------------------------------------- | :-------------------------------------------------- |
| **Falsy**| Acts like `False` | • Number `0`<br>• Empty text `""`<br>• `None` (nothing)         | Skips the `if` block and runs the `else` block instead. |
| **Truthy**| Acts like `True` | • Any number other than `0`<br>• Any text with letters (`"Shivam"`) | Runs the `if` block successfully.                   |

```python
## CODE EXAMPLE : ----------

username = "Shivam Kumar"

# We just pass the variable name directly without any comparison (like == "")
if username:
    # SCENARIO A (Truthy): The string has text inside, so Python treats it as True
    # and runs this block.
    print("Welcome back :", username)
else:
    # SCENARIO B (Falsy): If the string was empty (""), Python would treat it
    # as False and run this block instead.
    print("Guest user detected. Please log in.")
```
