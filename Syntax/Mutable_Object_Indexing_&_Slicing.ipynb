# **Indexing of mutable objects**

Each element in a mutable object is assigned a positional number. These positional numbers are also known as "**indices**." Using these positional numbers, we can **access**, **modify** the element within the mutable object.

**Important distinction:**
- **List**: Access elements using indices.
- **Dictionary**: Access values using keys.
- **Set**: No indexing or keys; use iteration to access elements.

### **Indexing For List**

There are **two** types of indexing:

1. **Positive Indexing** : Positive indexing starts with **0**

2. **Negative Indexing** : Negative indexing starts with **-1**

**Which mutable object allows indexing to work with their elements?** Only list

```python
# Indexing For List

l = [1,2,3,4,5]  # A List

## +ve Indexing

print(l[0])      # Accessing The First Element
print(l[1])      # Accessing The Second Element
print(l[2])      # Accessing The Third Element
print(l[3])      # Accessing The Fourth Element
print(l[4])      # Accessing The Fifth Element

## -ve Indexing

print(l[-5])      # Accessing The First Element
print(l[-4])      # Accessing The Second Element
print(l[-3])      # Accessing The Third Element
print(l[-2])      # Accessing The Fourth Element
print(l[-1])      # Accessing The Fifth Element
```

### **Indexing For Dictionary**

Python dictionaries don't support **positional indexing** like lists. They support **key-based access**.

```python
# Example

student = {
    "name": "Shivam",
    "age": 21,
    "course": "Python"
}

## 1. Using square brackets []

print(student["name"])    # Shivam
print(student["age"])    # 21
print(student["course"])  # Python

## 2. Using .get()

print(student.get("name"))  # Shivam
print(student.get("age"))   # 21

## 3(i). Using a for loop

for key in student:
    print(key)

## 3(ii). Access values

for value in student.values():
    print(value)

## 3(iii). Access Both

for key, value in student.items():
    print(key, value)
```

### **Indexing For Set**

Set does **not support** positional indexing like lists. Because they are unordered collections. If you try to use indexing like we did in the list, it will throw a **Type Error**. Then how can we access its elements?


```python
numbers = {10, 20, 30}

## 1. Using a for loop

for i in numbers:
    print(i)

## 2. Using iter() and next()

it = iter(numbers)
print(next(it))

## 3. Converting into a list

print(list(numbers)[0])
```

# **Modifying elements using indexing**

### **List**

```python
numbers = [10, 20, 30, 40, 50]

## 1. Modify using positive indexing

numbers[0] = 100
print(numbers)

## 2. Modify using negative indexing

numbers[-1] = 500
print(numbers)

## 3. Modify multiple elements individually

numbers[1] = 200
numbers[2] = 300
numbers[3] = 400

print(numbers)
```

### **Dictionary**

```python
## 1. Modifying dictionary values

student = {
  "name": "Rahul",
  "age": 20,
  "marks": 85
  }

student["age"] = 21
student["marks"] = 95

print(student)


## 2. Adding a new key-value pair

student["city"] = "Lucknow"
print(student)

## 3. Removing a key-value pair

del student["age"]
print(student)
```

### **Set**

```python
## 1. Adding an element using add()

s = {10, 20, 30}

s.add(40)
print(s)

## 2. Removing an element using remove()

s = {10, 20, 30}

s.remove(20)
print(s)

## 3. Removing an element using discard()

s = {10, 20, 30}

s.discard(50)                      # discard() doesn't raise an error if the element doesn't exist. `remove()` does.
print(s)

## 4. Updating multiple elements using update()

s = {10, 20, 30}

s.update([40, 50, 60])
print(s)
```

**Important distinction:**

| Operation               | List       | Dictionary      | Set                      |
| ----------------------- | ---------- | --------------- | ------------------------ |
| Modify existing element | Index      | Key             | Not directly             |
| Add element             | `append()` | Key assignment  | `add()`                  |
| Remove element          | `remove()` | `pop()` / `del` | `remove()` / `discard()` |
| Slicing                 | Yes        | No              | No                       |

## **Slicing**

### **List**

**Syntax**: sequence[start:stop:step]

- **start**: Starting index (included).
- **stop**: Ending index (excluded).
- **step**: How many positions to jump.

Lists support slicing because they are ordered and indexed.

```python
# Syntax

a = [10, 20, 30, 40, 50]

print(a[1:4])
print(a[:3])     # [10, 20, 30]
print(a[2:])     # [30, 40, 50]
print(a[::-1])   # [50, 40, 30, 20, 10]
print(a[::2])    # [10, 30, 50]

```

### **Dictionary**

```python
## Dictionaries use keys rather than numerical indices:
## You can convert dictionary items into a list and then slice:

d = {"a": 10, "b": 20, "c": 30}
print(list(d.items())[0:2])
```

### **Set**

```python
## Sets are unordered and do not have numerical indices.
# You can convert a set into a list and then slice it:

s = {10, 20, 30, 40, 50}
print(list(s)[1:4])



### Remember: set order is not guaranteed, so don't rely on a particular slicing result.

```
