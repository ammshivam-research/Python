# **BASIC Function**

### **1. input() – Taking input from the user**

**input()** is used to get data from the user while the program is running.

```python
# Basic version

name = input()
print(name)

# Program waits for the user to type something and press Enter.
# Whatever the user types is stored as a string.
```

```python
# With a message (most common)

name = input("Enter your name: ")
print("Hello", name)

# Shows a message to the user before waiting for input.
```

```python
# Taking a number (you must convert it)

age = int(input("Enter your age: "))
print(age + 5)

# input() always gives a string.
# Use int() or float() to convert it into a number.
```

```python
# Taking multiple values in one line

a, b = input("Enter two numbers: ").split()
print(a, b)
```

### **2. len() – Finding the length**

**len()** tells you how many items are inside something.

```python
# With a string

text = "Hello"
print(len(text))          # Output: 5
```

```python
# With a list

numbers = [10, 20, 30, 40]
print(len(numbers))       # Output: 4
```

```python
# With other things

print(len("Python"))      # 6
print(len([1, 2, 3]))     # 3
print(len((5, 6, 7, 8)))  # 4 (tuple)
```

### **3. print() – Showing output**

**print()** is used to display something on the screen.

```python
# Basic version

print("Hello World")
```

```python
# Printing variables

name = "Shivam"
age = 20
print(name)
print(age)
```

```python
# Printing multiple things together

print("Name:", name, "Age:", age)
```

```python
# Using sep (separator)

print("Apple", "Banana", "Mango", sep=" - ")
# Output: Apple - Banana - Mango
```

```python
# Using end (what comes at the end)

print("Hello", end=" ")
print("World")
# Output: Hello World   (stays on the same line)
```

```python
# Printing with f-string (modern & clean way)

name = "Shivam"
age = 20
print(f"My name is {name} and I am {age} years old.")
```

## **Key Concepts Explained: Mutability, Homogeneity, and Order**

### **1. Mutability (Mutable vs. Immutable)**

**What it means**: Whether an object’s internal value can be changed after it has been created, without creating a brand new object in computer memory.

**Mutable (Can change)**: You can modify, add, or delete items inside the same object.

**Example**: A list is mutable because you can change my_list[0] = 99 without changing where the list lives in memory.

**Immutable (Cannot change)**: Once created, its value is locked forever. If you try to modify it, Python actually creates a brand-new object behind the scenes.

**Example**: A string is immutable. If you change a string, Python replaces it with a completely new string in memory.

### **2. Homogeneity (Homogeneous vs. Heterogeneous)**

**What it means**: Whether a data structure can hold only one specific data type or a mixture of different data types.

**Homogeneous (Uniform / Same type)**: Every single element inside the collection must be of the exact same data type.

**Example**: A strict string is homogeneous because every element inside it is a character (str).

**Heterogeneous (Mixed / Different types)**: The collection can mix different types together freely.

**Example**: A Python list can hold an integer, a string, a boolean, and another list all at the same time: [42, "hello", True, [1, 2]].

### **3. Order (Ordered vs. Unordered)**

**What it means**: Whether the items in a data structure maintain a strict, predictable sequence where every item has a fixed position (index) that you can access.

**Ordered**: Items stay in the exact position you put them in, and you can access them using numbers like index 0, 1, 2.

**Examples**: Lists, tuples, and strings.

**Unordered**: The items do not have a guaranteed or fixed position. They might shift around in memory, meaning you cannot access them using standard numerical indices.

**Examples**: Sets (which rearrange items based on hash values for fast searching).

## **Python Data Type Conversion Guide: Transforming Between Types**

```python
# INTEGER TO FLOAT

num = int(input("Enter a number: "))
print(float(num))
print(type(float(num)))
```

```python
# INTEGER TO STRING

num = int(input("Enter a number: "))
print(str(num))
print(type(str(num)))
```

```python
# INT TO BOOLEAN

num = int(input("Enter a number: "))
print(bool(num))
print(type(bool(num)))

# INT TO BOOLEAN

num = int(input("Enter a number: "))
print(bool(num))
print(type(bool(num)))
```

```python
# INT TO LIST

num = int(input("Enter a number: "))
print(list(num))

# ERROR REASON: TypeError: 'int' object is not iterable

# Why: The list() function expects an iterable object.
# Iterable object meaning: When list() expects an iterable, it means you must give it something that contains multiple items (a sequence or collection) so it can break them apart and put them into a new list.
# An integer is a single scalar value and cannot be iterated over.
```

```python
# INT TO TUPLE

num = int(input("Enter a number: "))
print(tuple(num))

# ERROR REASON: TypeError: 'int' object is not iterable

# Why: The tuple() function expects an iterable object (like a string, list, or sequence).
# Iterable object meaning: When tuple() expects an iterable, it means you must give it something that contains multiple items (a sequence or collection) so it can break them apart and put them into a new list.
# Since an integer is a single scalar value, Python cannot iterate over it.

# HOW TO FIX:

# 1. To make a single-element tuple, use a trailing comma inside parentheses: (num,)
# 2. To get a tuple of individual digits, convert to string first: tuple(str(num))
```

```python
# Solution_1 :

print(tuple((num,)))
print(type(tuple((num,))))

# Solution_2 :

print(tuple(str(num)))
print(type(tuple(str(num))))

```
