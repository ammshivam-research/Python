## **While Loop**

## **Understand What a&#x20;**`while`**&#x20;Loop Is**

A `while` loop in Python is used to **execute a block of code repeatedly as long as a given condition remains&#x20;**`True`**.**

Let's understand this with a simple example.

Imagine you want to print numbers from 1 to 5.

```
i = 1
while i <= 5:    
print(i)    
i = i + 1
```

**Output:**

```
1
2
3
4
5
```

### **Understand the basic structure**

```
while condition:    # Code to execute
```

There are three important things:

1. `while` — tells Python to start a while loop.
2. `condition` — Python checks whether the condition is `True` or `False`.
3. Indented code — executes repeatedly as long as the condition is `True`.

```
i = 1
while i <= 5:    
print(i)
```

**Question:** What do you think will happen?

A. It prints numbers from 1 to 5.

B. It prints `1` only once.

C. It prints `1` infinitely.

**Answer:**



Correct! C

Let's understand exactly why.

Our code:

```
i = 1while i <= 5:    print(i)
```

**Step-by-step execution:**

| Iteration | Value of `i` | Condition `i <= 5` | Action            |
| --------- | ------------ | ------------------ | ----------------- |
| 1         | 1            | True               | Prints 1          |
| 2         | 1            | True               | Prints 1          |
| 3         | 1            | True               | Prints 1          |
| 4         | 1            | True               | Prints 1          |
| ...       | 1            | True               | Continues forever |

**Why does this happen?**

Because we never change the value of `i`.

Therefore:

- `i` remains 1.
- `1 <= 5` always evaluates to `True`.
- Python keeps executing the loop indefinitely.

This is called an **infinite loop**.

### **One important concept**

To prevent an infinite loop, we generally update the variable inside the loop.

For example:

```
i = 1
while i <= 5:    
print(i)    
i += 1
```

## **How Does Python Execute a&#x20;**`while`**&#x20;Loop?**

Let's take this example:

```
i = 1
while i <= 3:    
print(i)    
i += 1
```

Output:

```
1
2
3
```

Now let's understand exactly what Python does.

### **Step-by-step execution**

**Initialisation**

```
i = 1
```

We assign `1` to the variable `i`.

**Iteration 1**

Python checks:

```
1 <= 3
```

True

- Executes `print(i)` → prints `1`.
- Executes `i += 1` → changes `i` to `2`.

**Iteration 2**

Python checks:

```
2 <= 3
```

True

- Executes `print(i)` → prints `2`.
- Executes `i += 1` → changes `i` to `3`.

**Iteration 3**

Python checks:

```
3 <= 3
```

True

- Executes `print(i)` → prints `3`.
- Executes `i += 1` → changes `i` to `4`.

**Iteration 4**

Python checks:

```
4 <= 3
```

False

Python exits the loop. No more statements inside the loop are executed.

### **The most important thing to remember**

A `while` loop follows this cycle:

**1. Check the condition**

**2. If True, execute the code**

**3. Update the variable**

Return to Step 1

**If False → Exit the loop**

## **Understanding&#x20;**`while True`**&#x20;and&#x20;**`while False`

What happens when we write:

```
while True:    
print("Hello")
```

And what happens when we write:

```
while False:    
print("Hello")
```

We'll understand how Boolean values control loop execution.

**Question:** What do you think the first code will do?

A. Print Hello once.

B. Print Hello infinitely.

C. Print nothing.

Correct! **Answer B** 🎯

Let's understand why.

### **First code:&#x20;**`while True`

```
while True:    
print("Hello")
```

- `True` is a Boolean value.
- Python checks the condition.
- The condition is always `True`.
- Therefore, Python executes `print("Hello")`.
- Python checks the condition again.
- It remains `True`.

**Result:** `Hello` gets printed infinitely (until the program is interrupted).

### **Second code:&#x20;**`while False`

```
while False:    
print("Hello")
```

Here:

- The condition is immediately `False`.
- Python never enters the loop body.
- `print("Hello")` is never executed.

**Result:** Nothing gets printed.

## **Problem**

```
x = 1

if x < 4:    
print(x)  # Prints 1    
x += 1    # x becomes 2

if x < 4:    
print(x)  # Prints 2    
x += 1    # x becomes 3
```

Wait! Notice something important: there are **two separate&#x20;**`if`**&#x20;statements**.

Therefore, the actual output is:

```
1
2
```

So the correct answer is **B**, not A. My friend, this is a useful distinction!

Each `if` statement executes once, but both statements get their own opportunity to execute.

**Key takeaway:** Two separate `if` statements are not automatically a loop.

## **Using a&#x20;**`while`**&#x20;Loop to Calculate a Sum**

Now let's move from printing numbers to performing mathematical operations using a `while` loop.

Imagine you want to calculate the sum of all numbers from 1 to 5.

Mathematically:

\\[ 1+2+3+4+5=15 \\]

Let's implement this in Python.

```
i = 1
total = 0

while i <= 5:    
total = total + i    
i += 1

print(total)
```

**Output:**

```
15
```

### **1. Understand the variables**

We have two variables:

- `i = 1`: Controls how many times the loop executes.
- `total = 0`: Stores the accumulated sum.

### **2. Understand the execution**

| Iteration | i | total before addition | New total |
| --------- | - | --------------------- | --------- |
| 1         | 1 | 0                     | 1         |
| 2         | 2 | 1                     | 3         |
| 3         | 3 | 3                     | 6         |
| 4         | 4 | 6                     | 10        |
| 5         | 5 | 10                    | 15        |

When `i` becomes 6, the condition `i <= 5` becomes False, and the loop terminates.

### **3. Understand an important operation**

```
total = total + i
```

This means:

Take the existing value of `total`, add the current value of `i`, and store the result back in `total`.

This is called **accumulation**.

## **For Loop**


A `for` loop is used to **iterate over a sequence of elements**, executing a block of code for each element.

Let's take a simple example:

```
numbers = [10, 20, 30, 40]

for i in numbers:    
     print(i)
```

**Output:**

```
10
20
30
40
```

### **Understand the code**

1. `numbers` is a list containing four elements.
2. `for` tells Python to start iterating.
3. `i` is the loop variable.
4. `in` tells Python which iterable to iterate over.
5. `numbers` is the iterable.
6. `print(i)` prints the current element.

### **How does Python execute this loop?**

| Iteration | Value assigned to `i` | Output |
| --------- | --------------------- | ------ |
| 1         | 10                    | 10     |
| 2         | 20                    | 20     |
| 3         | 30                    | 30     |
| 4         | 40                    | 40     |

Notice something important:

**Python automatically takes the next element from the list and assigns it to&#x20;**`i`**.**

You don't need to write `i += 1` as you typically do in a `while` loop.

## **Step 2: Understanding How a&#x20;**`for`**&#x20;Loop Works Internally**

Let's use this example:

```
numbers = [10, 20, 30]for i in numbers:    print(i)
```

**Output:**

```
10
20
30
```

### **How does Python execute this code?**

Python follows these steps:

**Iteration 1**

- Takes the first element: `10`.
- Assigns it to `i`.
- Executes `print(i)`.
- Prints `10`.

**Iteration 2**

- Takes the second element: `20`.
- Assigns it to `i`.
- Executes `print(i)`.
- Prints `20`.

**Iteration 3**

- Takes the third element: `30`.
- Assigns it to `i`.
- Executes `print(i)`.
- Prints `30`.

**Termination**

Python checks whether another element is available.

There are no more elements, so the loop terminates.

### **Visualising the process**

**Original list**

# **10**

# **20**

# **30**

**Python takes one element at a time**

Iteration 1

**i = 10**

Iteration 2

**i = 20**

Iteration 3

**i = 30**

Loop terminates

### **Important distinction:&#x20;**`for`**&#x20;vs.&#x20;**`while`

| `for` loop                            | `while` loop                                        |
| ------------------------------------- | --------------------------------------------------- |
| Iterates over an iterable.            | Repeats while a condition is True.                  |
| Automatically gets the next element.  | You generally manage the variable updates yourself. |
| Stops when the iterable is exhausted. | Stops when its condition becomes False.             |
