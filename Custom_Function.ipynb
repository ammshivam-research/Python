## **CUSTOM FUNCTION STRUCTURE**

### Every **custom function** built by us is made up of **3 blocks**:

**1. Defining The Function**  

**2. Writing Input values**

**3. Managing The Output**

## **BLOCK - 1**

**DEFINING THE FUNCTION** :

To properly define the custom function, we need **2** Parts :            

1. Header
2. Body

### **Header**
```
def account(a, b):
```

The header is the first line when defining a function. It acts as the introduction or setup line of our custom function.

A header is made of four specific parts written in this exact order:

1. **The Keyword (`def`)**
* **What it is:** The word `def` (short for define).
* **What it does:** Tells Python that a new function is starting.


2. **The Function Name (`account`)**
* **What it is:** The custom label you give to your function.
* **What it does:** Gives your function a name so you can call and use it later in your program.

     **Note:** You can use any name you want, except the built-in function name, or custom function name that has already been made.

3. **The Parameters (`(a, b)`)**
* **What it is:** Variables inside parentheses.
* **What it does:** These are variables acting like an empty placeholders for the data you will send into the function later. (If the function needs no inputs, it is just empty parentheses `()`).


4. **The Colon (`:`)**
* **What it is:** A colon symbol at the very end of the line.
* **What it does:** Tells Python that the header setup is finished, and the indented steps of the function will begin on the next line.

### **Example Header**

```python
def account(a, b):

```

* **`def`** starts the function.
* **`account`** is the name of the custom function.
* **`(a, b)`** are the input placeholders.
* **`:`** ends the header line.

### **Body**

The body is the indented section of code that sits directly below the function header. It contains the actual instructions and math that the function performs using the input parameters.

A function body is made up of **2** main parts:

1. **The Execution Steps**
* **What it is:** The code lines inside the custom function that we build to do the work.
* **What it does:** It takes the values stores inside the parameters (like `a` and `b`), performs operations on them (like addition), and stores the result into the memory and assign that memory location a local variable (like `ans`).
* **Rule:** Every line in the body must be **indented** (usually by 4 spaces or 1 tab). This indentation tells Python that these lines belong inside the our custom function.


2. **The Return Statement (`return`)**
* **What it is:** A command starting with the word `return` followed by the value or variable you want to send back.
* **What it does:** It finishes the function's job, sends the final answer back out to the main program (or print statement), and shuts down the function's local workspace (discarding temporary variables like `a`, `b`, and `ans`).



### **Example Function Body**

```python
def account(a, b):
    # --- Start of the Body ---
    ans = a + b
    return ans
    # --- End of the Body ---
```

### **How Python Runs a Function?**

````python
def account(a,b):
    ans = a + b
    return ans

x = int(input())
y = int(input())

print("The final discount is: ", account(x,y))
````

First, Python sees your `def account` block and simply remembers how to do that math for later without running it yet. Next, it stops at `x` and `y` to grab those two numbers from the user and saves them in a memory location within RAM, and assign some variables name to that memory location. Then, when Python reaches the `print` line, it sees `account(x, y)` and jumps up to run your function, giving `a` and `b` the exact same numbers as `x` and `y`.

Inside the function, it adds those numbers together and saves the result in `ans`. Right after that, `return ans` hands that final number straight back into the `print` statement, swapping out `account(x, y)` with the actual number. Finally, `print()` takes that returned number and shows the whole sentence on your screen.

### **DOUBT 1 : Function Headers and Parameter Assignment**

**Why doesn't Python throw an error when we call account(x, y) even though the function is defined as account(a, b)?**

When you write `x = 5`, Python creates the number `5` in the computer's memory and attaches the label `x` to it. During a function call, passing `x` into `account(a,b)` causes Python to execute pass-by-assignment, attaching a second temporary label, `a`, to that exact same number `5` in memory. Consequently, `a` acts as a local variable within the function's scope to access the value, while `x` remains untouched outside. Once the function finishes running, the temporary label `a` is discarded, but `x` remains completely intact.



## **BLOCK - 2**

**WRITING INPUT VALUES :**

Earlier, we saw that `a` gets thrown away when the function finishes, but `x` stays safe outside.

This brings us to a major rule in Python called **scope**. Scope just means **"where a variable is allowed to live and be used."**

There are **two** main types:

1. **Local Variables (Inside the Function)**
* Variables like `a`, `b`, and `ans` live **only** inside the function.
* They are born when the function starts running, and they are completely destroyed the second the function hits `return` and finishes. If you try to use `ans` outside the function, Python will throw an error because it doesn't exist out there.

2. **Global Variables (Outside the Function)**
* Variables like `x` and `y` live in the main program outside any function.
* They can be used anywhere in your main code.

Think of it like a house: local variables live inside a locked bedroom (the function) and disappear when you leave the room, while global variables live in the living room where everyone can see them.

````python
# --- Global Scope (Outside) ---
x = 5
y = 10

def account(a, b):
    # --- Local Scope (Inside) ---
    ans = a + b
    return ans

# Calling the function using global variables
result = account(x, y)

# This works fine (Global Variables):
print("Result is:", result)

# THIS WILL CAUSE AN ERROR:
# print(ans)
# Python says: NameError: name 'ans' is not defined.
# Why? Because 'ans' is local and died when the function finished!
````

## **BLOCK - 3**

**MANAGING THE OUTPUT :**

Once the function finishes its job and the `return` statement throws the final answer back out to your main program, you have to decide what to do with that value. If you don't catch it or use it, Python will just throw it away!

Managing the output means choosing one of three things to do with that returned value:

### **1. Store it in a Variable**

You can catch the returned value and save it inside a new variable so you can use it later in your code.

````python
# 'result' catches the value returned by the function
result = account(5, 10)

# Now you can use 'result' whenever you want
print(result)
````

### **2. Print it Directly**

You can drop the returned value straight into a `print()` statement just like we did earlier, so it shows up on the screen right away.

````python
# The returned value goes straight into print()
print("The final answer is:", account(5, 10))
````

### **3. Use it in Math or Decisions**

You can use the returned value immediately inside a calculation or an `if` statement.

````python
# Using the returned value in a math problem
final_price = 100 - account(5, 10)
print(final_price)
````

Think of `return` as someone throwing a ball back to you. "Managing the output" is whether you catch it with your hands (save it to a variable), let it hit a target (print it), or kick it right back into play (use it in math).

## **MODULAR APPROACH**

A modular approach to programming means breaking down a large, complex problem into smaller, independent, and reusable functions (or modules). Each function is designed to handle one specific task.

### **Key Benefits :**

**Maintainability:** Easier to read, test, and update individual parts of the code without breaking the whole program.

**Reusability:** Functions written for one purpose can be reused across different parts of the program or in entirely different projects.

**Debugging:** When an error occurs, isolation makes it much easier to pinpoint which function is failing.

**Readability:** Code reads more like a high-level story rather than a dense wall of logic.
