## **Built-in Python Function**

Built-in Python Function comes pre-installed on our computer, rather than the function we write using `def`

## **What do you mean by 'calling a function'?**

Calling the function means telling the python to run and execute the function.

**But how would you call a built-in function in Python IDE?**

To call a built-in function in IDE, writing the function's name alone doesn't actually make it do anything. You have to "call" it by typing its name followed by parentheses `()` so Python knows it's time to perform the action.

**Here is what that looks like:**

**1. Just typing the name (Not calling it):**

```python
print  # Python just looks at the name, but does nothing.
```
**2. Calling the function:**

```python
print("Hello!")  # Adding the parentheses () "calls" or runs the print function!
```
### **SUMMARY**:

Think of a function name like a phone contact:

   * Knowing the name (like `print`) is just having the contact saved.
   * Putting parentheses `()` next to it is like **making the actual phone call** to wake it up and make it do its job!

## **Where does our "Built-in Python Function Comes From"?**

Our build-in python function comes from the **builtins module** that is originally stored inside the python installation folder in the HDD or SSD of our computer system written in C programming langauge.

## **How does this "Build-in Python Function" stored in HDD or SDD gets used?**

Whenever we start python, the **builtins module** stored inside the python installation folder within HDD or SSD of our computer system gets copied and loaded into the RAM or the primary memory of the computer. Once it gets loaded into the RAM, the functions inside the builtins module is ready to be used.

This is why you never have to import them manually (you don't need to write import print). They are already there.

#### **IMPORTANT AND INTERESTING POINT:**

1. Our python code gets executed through **Cpython interpreter**, which is a software written in C Programming langauge.




## **Build-in Python Function Anatomy:**

The official term for the anatomy or the layout of any build-in python function is called **'Function Signature'**.

Every built-in python function has its own function signature.

**Components of a function signature:**

1. **Function Name**
2. **Positional Arguments**
3. **Keyword Arguments**

| Component | is it required? | What is it? | Example in `print()` | What Happens If You Leave It Out? |
| :--- | :--- | :--- | :--- | :--- |
| **Function Name** | **Required** | The exact identifier or name you type to call the function into action. | `print` | Python throws a **`NameError`** because it doesn't know what you are trying to run. |
| **Positional Arguments** | **Required** <br>*(unless marked with `*`)* | Values passed into the function based strictly on their order. An asterisk (`*`) means you can pass multiple items. | `*objects`  <br>*(e.g., `"Hello", "World"`)*  | If a required positional argument is missing, Python throws a **`TypeError`** and tells you an argument is missing. |
| **Keyword Arguments** | **Optional** | Extra settings or configurations written with an `=` sign. They **always** have a built-in default value. | `sep=' '` or `end='\n'` | Python safely ignores you typing it and **automatically uses its preset default value** (e.g., a space or a new line). |

### **NOTE:**

**Order Matters:**

1. **Positional arguments** must always come first in the function signature.
2. **Keyword arguments** must always come last in the function signature.
        
      **The `=` Sign Rule:**
      
      If you see an `=` sign in a function layout, it means that argument is optional because Python already has a backup value or default value ready to use.

````python
## CODE EXAMPLE:-----------------------------------------------------------

# 1. 'Hello' and 'World' are required positional arguments.
# 2.  sep='-' is an optional keyword argument overriding the default space.

print("Hello", "World", sep="-")

### Output: Hello-World
````

## **How Python Executes Code?**

When you type your code and hit run in a Python IDE, the **CPython interpreter**, software created by developers using the C programming language, acts as a multi-stage translator.

The moment you type English characters like `p`, `r`, `i`, `n`, `t`, your keyboard and operating system instantly convert them into standard character numbers, which are represented by binary patterns of electrical pulses (zeros and ones).

CPython reads these character patterns, breaks them into small puzzle pieces called **tokens**, and arranges those tokens into an **Abstract Syntax Tree (AST)** to check your grammar and ensure your code structure is completely valid.

Next, CPython converts the AST into an intermediate format known as **bytecode**. Unlike human-readable code or physical machine code, bytecode is simply a series of quick instruction ID numbers (called opcodes) created specifically for Python. Because your computer processor cannot read Python bytecode directly, it needs a middleman to translate those ID numbers into real action.

This is where the **Python Virtual Machine (PVM)** takes over. The PVM itself along with the rest of the CPython interpreter was written in C. However, before you ever installed Python, its developers passed all that C code through a C compiler, translating the entire interpreter once and for all into pure binary machine code (zeros and ones). So when you run Python, the interpreter is already a binary application running directly on your CPU.

As the PVM reads your bytecode ID numbers one by one, it matches each ID against an internal lookup list built into its pre-compiled code. Finding a match simply means taking a bytecode ID number and instantly executing the corresponding block of binary machine code that came from the original C routines (pre-written tasks).

These binary instructions send physical bursts of electricity through your computer's circuits, talking directly to your operating system and hardware to bring your high-level code to life on the screen.

## **Print() Function:**




### **What the print() Function Actually Does:**

**Displays Output on the Screen:** Its most common job is taking text or numbers and rendering them onto your monitor so you can check your program's results, debug errors, or show information to a user.

**Converts Things to Text:** Even if you give it numbers (42), booleans (True), or math equations, print() automatically converts them into readable text characters before showing them.

**Packages and Formats Data:** As shown in your notes, it can combine multiple items, insert custom separators (sep="..."), change how lines end (end="..."), build dynamic sentences using f-strings.

Think of print() like a megaphone for your code: whenever your program needs to show you what it is thinking or doing, it "speaks" through print().

### **There are many ways we can use print() function:**

1. **Basic Printing**
2. **Customizing Seperators using print() function**
3. **Customizing End Characters using print() function**
4. **Formatted Strings (f-strings)**


1. **Basic Printing:**

    You can print strings, numbers, and boolean values directly. Separate multiple items with commas, and Python will automatically put a space between them.

````python
# CODE EXAMPLE :-----------

print("Hello, World!")  # Printing a string
print(42)               # Printing an integer
print(True)             # Printing a boolean

# Printing multiple items (comma automatically adds spaces)
print("Score:", 95)     # Output: Score: 95
````

2. **Customizing Seperators using print() function**

    By default, print() separates multiple items with a space. You can change this using the sep parameter to specify any character or string you want between items.


````python
## CODE EXAMPLE :-----------

# Using a hyphen instead of a space
print("2026", "10", "31", sep="-")  # Output: 2026-10-31

# Using no separator
print("Py", "thon", sep="")         # Output: Python
````

3. **Customizing End Characters using print() function**

    By default, print() automatically adds a new line (\n) at the very end, which is why consecutive print statements appear on new lines. You can change the end parameter to change what happens at the end of the print statement.


````python
## CODE EXAMPLE:-------------------

print("Hello", end=" ")
print("World!")  # Output: Hello World! (printed on the same line)
````

4. **Formatted Strings (f-strings**)

    Introduced in Python 3.6, f-strings let you embed variables and expressions directly inside a string by placing an f before the opening quote and wrapping variables in curly braces {}.

````python
name = "Alice"
age = 25

# Directly embedding variables
print(f"My name is {name} and I am {age} years old.")
# Output: My name is Alice and I am 25 years old.

# You can even do math inside the braces!
print(f"Next year, I will be {age + 1} years old.")
# Output: Next year, I will be 26 years old.
````

## **Input() Function**

### **What the input() Function Actually Does:**

**Pauses the Program:** When you hit an input() line, the line freezes completely. The program refuses to move forward to the next line of code until the user types something on their keyboard and presses Enter.

**Captures Text**: Whatever the user types is captured by Python as a raw text string.

**Allows Custom Prompts**: You can pass a string inside the parentheses (like input("Enter your name: ")) to act as a friendly question or instruction so the user knows what they are supposed to type.

**Returns Data for Your Code to Use**: Once the user presses Enter, input() hands that typed text back to your program so you can save it into a variable, check it, do math with it (after type casting), or save it to a file.

```Think of input() like an interactive question on a form: the program stops, looks at you, and waits patiently for your answer before it can finish its job!```

### **There are many ways we can use Input() Function**

1. **Basic Input (Without a Prompt)**
2. **Input with a Prompt Message**
3. **Converting Input Type (Type Casting)**
4. **Taking Multiple Inputs in a Single Line**

**1. Basic Input (Without a Prompt)**

Using input() function will allow user to type something. The next line of the program will not be executed until the user type something.

````python
# The program waits for input silently

user_input = input()
print("You typed:", user_input)
````

**2.** **Input with a Prompt Message**

**Meaning** : You can write some string within the parentheses of input() to let the user know what he needs to type when the input syntax will be executed.

````python
name = input("Enter your name: ")
print(f"Hello, {name}!")
````

**3. Converting Input Type (Type Casting)**

Because input() always returns a string, if you want the user to type a number (like an age or a score), you must wrap the input function inside int() or float().

````python
# Taking an integer input
age = int(input("Enter your age: "))

# Now you can do math with it
print(f"Next year, you will be {age + 1} years old.")
````

**4. Taking Multiple Inputs in a Single Line**

Using the **.split() function** with the input() function , we allow the user to type multiple values separated by spaces. and then potentially assign them to multiple variables at once (combining multiple assignment!).

````python
name, age = input("Enter your name and age separated by a space: ").split()

print("Name:", name)
print("Age:", age)
````

## **Type() Function**


### **What the type() Function Actually Does:**

**Inspects Data Types:** You pass any value or variable into **type()**, and it inspects it to tell you whether Python sees it as an integer (int), a decimal (float), text (str), a boolean (bool), or something else.

**Helps with Debugging:** When your code crashes because you tried to add text to a number, **type()** lets you check what kind of data you are actually assigned to your variables.

````python
# Checking integers
print(type(42))            # Output: <class 'int'>

# Checking text strings
print(type("Hello"))       # Output: <class 'str'>

# Checking decimals (floats)
print(type(3.14))          # Output: <class 'float'>

# Checking booleans
print(type(True))          # Output: <class 'bool'>

# Checking a variable
age = 25
print(type(age))           # Output: <class 'int'>
````

| Type Name | Short Description | Example & `type()` Output |
| :--- | :--- | :--- |
| `int` | Whole numbers without any decimal points. | `type(42)` -> `<class 'int'>` |
| `float` | Numbers that contain a fractional or decimal component. | `type(3.14)` -> `<class 'float'>` |
| `str` | Text characters enclosed within quotes. | `type("Hello")` -> `<class 'str'>` |
| `list` | An ordered, changeable collection of items inside square brackets `[]`. | `type([1, 2, 3])` -> `<class 'list'>` |
| `tuple` | An ordered, unchangeable collection of items inside parentheses `()`. | `type((1, 2, 3))` -> `<class 'tuple'>` |
| `dict` | A collection of key-value pairs stored inside curly braces `{}`. | `type({"name": "Alice"})` -> `<class 'dict'>` |
| `set` | An unordered collection of unique items that automatically removes duplicates. | `type({1, 2, 3})` -> `<class 'set'>` |
| `bool` | Binary truth values restricted strictly to `True` or `False`. | `type(True)` -> `<class 'bool'>` |
| `NoneType` | Represents the intentional absence of a value (`None`). | `type(None)` -> `<class 'NoneType'>` |

## **The id() Function:**


Using **id() function**, we can find out the **memory address** of any **PyObject** in decimal format.

**What it does**: When you type any kind of data with your keyboard (like a number, string, or list), python stores that data into a PyObject and then it stores that PyObject within a specific section of the electrical circuits, which is a part of memory array banks within the memory chips of RAM, in the electrical form.

and each specific section of that electrical circuits which is a part of memory array banks can be recognized with a memory address.

The id() function shows that memory address where our PyObject is stored.

**The Rule**: Every object in Python has its own unique ID while it exists. If two variables point to the exact same object, they will share the exact same id().

````python
## Same Address

x = 42
print(id(x))  # Output: Address in decimal format !

y = x
print(id(y))  # Output: Same exact number as id(x), because they point to the same PyObject!
````

````python
## Same Address

x1 = 76
print(id(x1))   # Output: Address in decimal format !

y1 = 76
print(id(y1))   # Output: Same exact number as id(x), because they point to the same PyObject!
````

````python
## Different Address

x2 = 67
print(id(x2))   # Output: Address in decimal format !

y2 = 98
print(id(y2))   # Output: Different address in decimal format !
````

## **The hex() Function: Converting IDs to Hexadecimal**

Using the **hex() function** will help you to translate the decimal number format into hexadecimal number format.

**What it does:** hex() function takes any regular decimal number format and converts it into a base-16 hexadecimal string format (starting with 0x).

**Why it's used with id()**: Computer scientists and programmers usually like looking at memory addresses in hexadecimal (base-16) instead of massive regular integers because it's easier to read.

You will often see the combined syntax like this: **hex(id(obj))**

````python
x = 42
memory_address = id(x)

print(memory_address)       # Output: (regular integer)
print(hex(memory_address))  # Output: (hexadecimal memory address)

# Or all in one line:
print(hex(id(x)))           # Directly shows the memory address in hexadecimal format
````

## **The .split() Function: Chopping Text into Pieces**

Using the **.split() function**, we can cut the sentence into distinct words.

**What it does**: It takes a single string of text, looks for a specific separating character (called a **delimiter**), chops the text apart at those points, and packages all the resulting pieces into a list.

**Default behavior**: If you don't give it any instructions inside the parentheses (.split()), it automatically splits the text wherever it finds spaces.

````python
# Basic split (splits by spaces)
sentence = "Python is awesome"
words = sentence.split()

print(words)  # Output: ['Python', 'is', 'awesome'] (a list of strings)


# Custom split (splitting by a specific character like a comma)
csv_data = "Alice,25,Engineer"
details = csv_data.split(",")

print(details)  # Output: ['Alice', '25', 'Engineer']
````

## **len() : Finding The Length**

| Data Type | Category | Supports `len()`? | Example & Behavior |
| :--- | :--- | :--- | :--- |
| **Integer (`int`)** | Scalar (single value) | ❌ No | `len(42)` $\rightarrow$ TypeError |
| **Float (`float`)** | Scalar (single value) | ❌ No | `len(3.14)` $\rightarrow$ TypeError |
| **String (`str`)** | Container (collection of characters) | ✅ Yes | `len("hello")` $\rightarrow$ 5 |
| **Boolean (`bool`)** | Scalar (single value) | ❌ No | `len(True)` $\rightarrow$ TypeError |
| **List (`list`)** | Container (collection of items) | ✅ Yes | `len(['a', 'b'])` $\rightarrow$ 2 |
| **Tuple (`tuple`)** | Container (collection of items) | ✅ Yes | `len((1, 2))` $\rightarrow$ 2 |
| **Set (`set`)** | Container (collection of unique items) | ✅ Yes | `len({1, 2})` $\rightarrow$ 2 |
| **Dictionary (`dict`)** | Container (collection of key-value pairs) | ✅ Yes | `len({'a': 1})` $\rightarrow$ 1 |
| **None (`none`)** | Scalar (single null value) | ❌ No | `len(None)` $\rightarrow$ TypeError |

````python
# --- Containers (Support len()) ---

# 1. String (str)
text = "hello"
print("String length:", len(text))  # Output: 5

# 2. List (list)
my_list = ['a', 'b']
print("List length:", len(my_list))  # Output: 2

# 3. Tuple (tuple)
my_tuple = (1, 2)
print("Tuple length:", len(my_tuple))  # Output: 2

# 4. Set (set)
my_set = {1, 2}
print("Set length:", len(my_set))  # Output: 2

# 5. Dictionary (dict) - Counts key-value pairs
my_dict = {'a': 1}
print("Dictionary length:", len(my_dict))  # Output: 1


print("-" * 40)


# --- Scalars (Do NOT support len() - will raise TypeError) ---

# We use try-except blocks so the code runs without crashing completely

# 6. Integer (int)
try:
    print(len(42))
except TypeError as e:
    print("Integer error:", e)

# 7. Float (float)
try:
    print(len(3.14))
except TypeError as e:
    print("Float error:", e)

# 8. Boolean (bool)
try:
    print(len(True))
except TypeError as e:
    print("Boolean error:", e)

# 9. None (none)
try:
    print(len(None))
except TypeError as e:
    print("None error:", e)
````
